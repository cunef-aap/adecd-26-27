## Celdas: ejecutar y estado

Un cuaderno es una lista de **celdas**. Las de código se ejecutan y las de texto solo se leen.

- Para ejecutar una celda: **Mayús + Intro**. Ejecuta y salta a la siguiente.
- El número entre corchetes a la izquierda, `[7]`, dice que esa celda fue la séptima que se ejecutó. Es el **orden de ejecución**, no el orden en que está escrita en la página.
- Las celdas **comparten estado**: una variable definida en una celda sigue existiendo en las siguientes. Por eso el orden importa.
- Cuando algo deja de tener sentido: **Entorno de ejecución → Reiniciar y ejecutar todo**. Eso borra todas las variables y vuelve a ejecutar de arriba abajo, que es el estado limpio.

Esa última opción resuelve la mayoría de los problemas del primer día.

## Variables

Una variable es un nombre para un valor. Se asigna con `=`, que no es igualdad matemática sino “guarda esto con este nombre”.

```python
n = 20
sigma = 3.0
titulo = "señal más ruido"
print(n, sigma, titulo)
```

`20` es un entero y `3.0` un decimal. Los textos van entre comillas. La última expresión de una celda se muestra sola, sin `print`:

```python
n * sigma
```

## Listas, tuplas y desempaquetado

Una lista se escribe entre corchetes y se indexa **desde cero**:

```python
pesos = [0.2, 0.2, 0.6]
pesos[0], pesos[2], len(pesos)
```

Los paréntesis hacen una tupla, que funciona igual pero no se puede modificar. Y una asignación puede repartir varios valores a la vez, algo que el curso usa mucho:

```python
a, b = -0.3, 0.2
media, desviacion = 0.0, 1.0
a, b, media, desviacion
```

## Funciones

`def` define una función: un cálculo con nombre, que recibe argumentos y **devuelve** un valor con `return`. El cuerpo va indentado cuatro espacios, y la indentación es obligatoria: es lo que marca dónde empieza y acaba la función.

```python
def recta(x, pendiente, intercepto):
    return pendiente * x + intercepto

recta(2.0, 3.0, 1.0)
```

Los argumentos se pueden pasar **por nombre**, y así se lee mucho mejor cuando son varios:

```python
recta(2.0, intercepto=1.0, pendiente=3.0)
```

## Bucles y condiciones

`for` repite algo para cada elemento de una lista, y `if` decide. Los dos usan indentación.

```python
mejor = -100.0
for candidato in [1.5, 2.0, 2.5]:
    valor = -(candidato - 2.0) ** 2
    if valor > mejor:
        mejor = valor
        mejor_candidato = candidato

mejor_candidato, mejor
```

Ese bucle es, en pequeño, el del final del capítulo 1: probar candidatos y quedarse con el mejor.

## Tensores: los vectores del curso

Una parte importante del curso trabaja con **tensores** de `torch`. Un tensor es una tira de números sobre la que las operaciones actúan **elemento a elemento**, sin escribir bucles.

```python
import torch

x = torch.tensor([0.0, 1.0, 2.0, 3.0])
2 * x + 1
```

Compáralo con una lista: `2 * [0.0, 1.0]` repetiría la lista en lugar de multiplicar. Las tres formas de crear tensores que usa el capítulo 1:

```python
torch.linspace(0, 10, 5)        # 5 valores igualmente espaciados de 0 a 10
```

```python
torch.randn(4)                  # 4 números al azar de una normal estándar
```

```python
torch.normal(3.0, 2.0, size=(4,))   # al azar, con media 3 y desviación 2
```

Dos métodos que aparecen en muchas celdas:

- **`.item()`** saca el número de un tensor de **un solo** elemento, para poder usarlo como un decimal normal. Sobre un tensor de varios da error.
- **`.numpy()`** convierte el tensor en un array de `numpy`, que es lo que entiende `matplotlib` al dibujar.

```python
y = torch.tensor([1.0, 4.0, 9.0])
y[0].item(), y.numpy()
```

## Seleccionar partes de un tensor

Existen varias formas de indexar tensores:

1.  Por **posición**, con un índice o un par `desde:hasta`, donde el segundo extremo no entra:

```python
x = torch.linspace(0, 10, 8)
x[0], x[-1], x[:3], x[2:5]
```

1.  Por **condición**, con una máscara de verdadero y falso. La comparación se aplica elemento a elemento y devuelve otro tensor:

```python
grandes = x > 5
grandes, x[grandes], grandes.sum().item()
```

Y por **lista de índices**:

```python
indices = torch.tensor([0, 2, 4])
x[indices]
```

## Dos dimensiones: la matriz de datos

A partir del capítulo 3 los datos no son una tira sino una **tabla**: una fila por observación y una columna por variable. En código es un tensor de dos dimensiones.

```python
X = torch.stack([torch.ones(5), torch.linspace(0, 4, 5)], dim=1)
X
```

Aquí, `dim` indica que ambos vectores han de juntarse por columnas.

`.shape` da el tamaño como (filas, columnas). Mirar la forma de un tensor es el primer diagnóstico cuando algo no cuadra:

```python
X.shape
```

Se selecciona con dos índices separados por coma: primero filas, después columnas. Los dos puntos solos significan “todas”:

```python
X[0], X[:, 1]
```

Y `@` es el producto matricial.

```python
w = torch.tensor([1.0, 2.0])
X @ w
```

## Resumir y mostrar números

Los resúmenes estadísticos habituales, no requieren bucles.

```python
y = torch.tensor([2.0, 4.0, 4.0, 5.0])
y.mean(), y.std(), y.sum(), len(y)
```

Los resultados del curso son números que hay que leer, así que conviene mostrarlos con los decimales justos. La `f` delante de las comillas permite meter un cálculo entre llaves, y `:.4f` pide cuatro decimales:

```python
error = ((y - y.mean()) ** 2).mean()
print(f"error cuadrático medio: {error.item():.4f}")
```

## Leer datos con pandas

Los conjuntos de datos del curso están publicados, así que se leen directamente desde su dirección, sin descargar nada a mano. La librería que usaremos para la lectura es `pandas`. Los objetos con los que trabaja se denominan **DataFrame**: una tabla con nombres de columna.

```python
import pandas as pd

url = "https://cunef-aap.github.io/adecd-26-27/datos/airbnb_madrid.csv"
datos = pd.read_csv(url)
datos.shape
```

Quince mil filas y diecisiete columnas. `.head(3)` muestra las tres primeras:

```python
datos.head(3)
```

Una columna se saca por su nombre entre corchetes, y puede resumirse igual que un tensor:

```python
datos["price"].mean()
```

Para pasar de una columna de `pandas` a un tensor de `torch`:

```python
precio = torch.tensor(datos["price"].values, dtype=torch.float32)
precio[:5]
```

Para llevarse **varias columnas a la vez**, se pasa la lista de nombres. El resultado sigue siendo una tabla, y `.dropna()` quita las filas a las que les falte algún valor:

```python
columnas = ["accommodates", "bedrooms", "number_of_reviews"]
tabla = datos[columnas].dropna()
tabla.shape
```

Y esa tabla se convierte en una matriz de `torch` igual que antes, con `.values`. Es exactamente la matriz de datos del capítulo 3: una fila por caso y una columna por variable.

```python
X = torch.tensor(tabla.values, dtype=torch.float32)
X.shape, X[:3]
```

Hay una segunda forma de seleccionar, por **posición** en lugar de por nombre, que es `.iloc`. Lleva dos índices separados por coma, filas y columnas, igual que un tensor de dos dimensiones:

```python
datos.iloc[0:3, 0:4]
```

Entre las dos, en el curso se usa casi siempre la selección por nombre: es más larga de escribir pero no se rompe cuando cambia el orden de las columnas.

## Dibujar una figura

Para dibujar gráficos, usaremos matplotlib.

```python
from matplotlib import pyplot as plt

x = torch.linspace(0, 10, 20)
y = 2 * x + 1 + 3 * torch.randn(20)

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x.numpy(), y.numpy(), color="black", facecolors="none",
           label="datos")
ax.plot(x.numpy(), (2 * x + 1).numpy(), color="#ff5700", label="señal")
ax.set(xlabel="$x$", ylabel="$y$")
ax.legend()
plt.tight_layout()
```

`ax.scatter` pinta puntos y `ax.plot` una línea. Fíjate en los `.numpy()`: sin ellos `matplotlib` no sabe dibujar un tensor.

## La semilla

`torch.randn` da números distintos cada vez, así que una figura no saldría dos veces igual. Fijar la **semilla** hace que la secuencia de números al azar sea siempre la misma:

```python
torch.manual_seed(42)
primera = torch.randn(3)
torch.manual_seed(42)
segunda = torch.randn(3)
primera, segunda
```

Los dos tensores coinciden. Todos los capítulos fijan la semilla al principio, y de ahí que las cifras del texto sean reproducibles.

## Los cinco errores del primer día

| Lo que sale en rojo | Qué ha pasado |
|------------------------------------|------------------------------------|
| `NameError: name 'x' is not defined` | no has ejecutado la celda donde se define `x`. Ejecuta de arriba abajo |
| `IndentationError` | falta o sobra la indentación del cuerpo de un `def`, un `for` o un `if` |
| `TypeError: only one element tensors can be converted...` | `.item()` sobre un tensor de varios elementos |
| `RuntimeError: ... size mismatch` o `shapes cannot be multiplied` | un `@` entre tensores cuyas formas no encajan. Mira los dos `.shape` |
| Ningún error, pero el número no cuadra | cambiaste una celda de arriba y no la volviste a ejecutar. Reiniciar y ejecutar todo |

El último es el más traicionero, porque no avisa. Ante cualquier resultado que no cuadre, lo primero es reiniciar y ejecutar todo.
