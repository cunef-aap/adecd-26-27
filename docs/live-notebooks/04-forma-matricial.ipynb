# Forma matricial y ecuaciones normales

Una sola identidad, $\mathbf{v}^{\mathsf{T}}\mathbf{X}^{\mathsf{T}}\mathbf{X}\mathbf{v} = \lVert \mathbf{X}\mathbf{v}\rVert^2$, sostiene el capítulo entero

## Por qué estamos aquí

Con una variable ya sabemos minimizar. Con $p$ variables hace falta álgebra matricial, y a cambio obtenemos algo que con una variable no se veía: **una fórmula cerrada**, y la respuesta exacta a cuándo existe y cuándo es única.

## Preparativos

<span class="theorem-title">**Definición 1 (Matriz de diseño) **</span>$\mathbf{X}\in \mathbb{R}^{n\times (p+1)}$ con **primera columna de unos**, de modo que $x_{i0} = 1$ y el término independiente queda absorbido en $\boldsymbol{w}$. ## Ecuaciones normales

<span class="theorem-title">**Teorema 1 (Gradiente del riesgo) **</span>$\displaystyle \nabla\hat{R}(\boldsymbol{w}) = \frac{2}{n}\,\mathbf{X}^{\mathsf{T}}(\mathbf{X}\boldsymbol{w}- \mathbf{y})$.

<span class="proof-title">*Prueba*. </span>

$\square$

<span class="theorem-title">**Teorema 2 (Ecuaciones normales) **</span>Si $\hat{\boldsymbol{w}}$ minimiza $\hat{R}$, entonces $\mathbf{X}^{\mathsf{T}}\mathbf{X}\hat{\boldsymbol{w}}= \mathbf{X}^{\mathsf{T}}\mathbf{y}$.

<span class="proof-title">*Prueba*. </span>**?@thm-condicion-primer-orden** aplicado a [Teorema 1](#thm-gradiente-matricial): en un mínimo el gradiente se anula, y $\frac{2}{n}\mathbf{X}^{\mathsf{T}}(\mathbf{X}\hat{\boldsymbol{w}}- \mathbf{y}) = \mathbf{0}$ equivale a $\mathbf{X}^{\mathsf{T}}\mathbf{X}\hat{\boldsymbol{w}}= \mathbf{X}^{\mathsf{T}}\mathbf{y}$. $\square$

<span class="theorem-title">**Corolario 1 (Ortogonalidad de los residuos) **</span>Con $\mathbf{r}= \mathbf{y}- \mathbf{X}\hat{\boldsymbol{w}}$ se tiene $\mathbf{X}^{\mathsf{T}}\mathbf{r}= \mathbf{0}$. En particular, **si hay término independiente**, $\sum_i r_i= 0$ y la media de las predicciones coincide con $\bar{y}$.

<span class="proof-title">*Prueba*. </span>Reescritura de [Teorema 2](#thm-ecuaciones-normales). La primera componente de $\mathbf{X}^{\mathsf{T}}\mathbf{r}$ es $\mathbf{1}^{\mathsf{T}}\mathbf{r}= \sum_i r_i$ porque la primera columna de $\mathbf{X}$ es de unos. $\square$

> **Important**
>
> [Corolario 1](#cor-residuos-ortogonales) parece un tecnicismo y es la pieza que hace funcionar **?@thm-sct-descompone** en el capítulo 6 — y, al fallar fuera de la muestra, la que explica por qué $R^2$ puede salir negativo.

## Óptimo global, existencia y unicidad

<span class="theorem-title">**Teorema 3 (Identidad fundamental) **</span>Si $\hat{\boldsymbol{w}}$ resuelve [Teorema 2](#thm-ecuaciones-normales), entonces **para todo** $\boldsymbol{w}$: $$
\hat{R}(\boldsymbol{w}) = \hat{R}(\hat{\boldsymbol{w}}) + \frac{1}{n}\left\lVert \mathbf{X}(\boldsymbol{w}- \hat{\boldsymbol{w}}) \right\rVert^2 .
 \qquad(1)$$

<span class="proof-title">*Prueba*. </span>

$\square$

<span class="theorem-title">**Corolario 2 (Todo punto crítico es mínimo global) **</span>El segundo sumando de [Ecuación 1](#eq-identidad) es $\geq 0$, luego $\hat{R}(\boldsymbol{w}) \geq \hat{R}(\hat{\boldsymbol{w}})$ para todo $\boldsymbol{w}$.

> **Important**
>
> Fíjense en lo que acaba de pasar: hemos demostrado **optimalidad global sin una sola palabra sobre convexidad**. Sumar y restar ha hecho todo el trabajo.

<span class="theorem-title">**Teorema 4 (Existencia y unicidad) **</span>$\mathop{\mathrm{nuc}}(\mathbf{X}^{\mathsf{T}}\mathbf{X}) = \mathop{\mathrm{nuc}}(\mathbf{X})$. En consecuencia, $\hat{\boldsymbol{w}}$ es **único** si y solo si ninguna columna de $\mathbf{X}$ es combinación lineal de las demás (es decir $\mathop{\mathrm{rango}}\mathbf{X}= p+ 1$, lo que exige $n\geq p+ 1$), y entonces $$\hat{\boldsymbol{w}}= (\mathbf{X}^{\mathsf{T}}\mathbf{X})^{-1}\mathbf{X}^{\mathsf{T}}\mathbf{y}.$$

<span class="proof-title">*Prueba*. </span>

$\square$

<span class="theorem-title">**Teorema 5 (Hessiano y convexidad) **</span>$\nabla^2\hat{R}= \frac{2}{n}\mathbf{X}^{\mathsf{T}}\mathbf{X}$. Para cualquier dirección $\mathbf{v}$, la función $\varphi(t) = \hat{R}(\boldsymbol{w}+ t\mathbf{v})$ cumple $\varphi''(t) = \frac{2}{n}\left\lVert \mathbf{X}\mathbf{v} \right\rVert^2 \geq 0$: **la curvatura en toda dirección es no negativa**, es decir $\hat{R}$ es convexa. Es estrictamente convexa si y solo si $\mathop{\mathrm{rango}}\mathbf{X}= p+1$.

<span class="proof-title">*Prueba*. </span>Derivar [Teorema 1](#thm-gradiente-matricial) con [Lema 1](#lem-grad-lineal) y [Lema 2](#lem-grad-cuadratica); la curvatura direccional es la misma línea $\left\lVert \mathbf{X}\mathbf{v} \right\rVert^2$ de [Teorema 4](#thm-existencia-unicidad). $\square$

> **Note**
>
> Orden deliberado: [Teorema 3](#thm-identidad-fundamental) da el mínimo global **primero**, y el hessiano llega después como segunda lectura. Al revés, la convexidad costaría media pizarra de teoría de funciones convexas; así cuesta tres líneas. El hessiano se queda porque es lo que generaliza a *ridge* (capítulo 8) y a Newton.

## Coste computacional

<span class="theorem-title">**Proposición 2 (¿Fórmula cerrada o gradiente?) **</span>Formar $\mathbf{X}^{\mathsf{T}}\mathbf{X}$ cuesta $O(np^2)$; resolver el sistema, $O(p^3)$. Un paso de descenso cuesta $O(np)$, y $T$ pasos $O(Tnp)$. **El gradiente gana cuando $p^2 \gg Tn$**, o cuando $\mathbf{X}$ no cabe en memoria, o cuando los datos llegan en flujo.

## Nunca invertir la matriz

Se resuelve el sistema (`solve`, `lstsq`), no se calcula la inversa. El número de condición cumple $\kappa(\mathbf{X}^{\mathsf{T}}\mathbf{X}) = \kappa(\mathbf{X})^2$, que **enunciamos y no demostramos**: en el laboratorio se ve numéricamente con una matriz casi colineal.

## La conversación pendiente con econometría

## Aquí declinamos formalmente la inferencia

En este punto la tentación es calcular errores estándar, estadísticos $t$ y valores $p$ sobre $\hat{\boldsymbol{w}}$. **No lo vamos a hacer**, y conviene decir por qué en lugar de omitirlo en silencio. Toda esa maquinaria responde a la pregunta *¿cuánto vale $w_j$?*; este curso responde a *¿qué predice mejor fuera de la muestra?*. Son objetivos distintos y llevan a modelos distintos: la regularización del capítulo 8 sesga deliberadamente $\hat{\boldsymbol{w}}$ porque a cambio reduce el error de predicción.

La deuda no queda impagada: en la semana 15 daremos incertidumbre sobre las **predicciones** mediante intervalos conformes, sin un solo valor $p$.

## La trampa de esta semana

## `np.linalg.inv(X.T @ X) @ X.T @ y`

Funciona en clase y falla con datos reales. Dos motivos: si alguna columna es combinación de las demás, la inversa **no existe** ([Teorema 4](#thm-existencia-unicidad)) y NumPy devuelve basura numérica en lugar de un error; y aunque exista, elevar al cuadrado el número de condición pierde precisión. Use `solve` o `lstsq`.

## Qué llevarte

-   $\mathbf{X}^{\mathsf{T}}\mathbf{X}\hat{\boldsymbol{w}}= \mathbf{X}^{\mathsf{T}}\mathbf{y}$ sale de anular el gradiente ([Teorema 2](#thm-ecuaciones-normales)).
-   La identidad [Ecuación 1](#eq-identidad) da el **mínimo global** sin teoría de convexidad, y de regalo la unicidad ([Teorema 4](#thm-existencia-unicidad)).
-   Todo se apoya en una línea: $\mathbf{v}^{\mathsf{T}}\mathbf{X}^{\mathsf{T}}\mathbf{X}\mathbf{v} = \left\lVert \mathbf{X}\mathbf{v} \right\rVert^2$.
