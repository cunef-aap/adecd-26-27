# Forma matricial y ecuaciones normales

Ecuaciones normales, optimalidad global, existencia y unicidad

## Planteamiento

Con una variable ya sabemos minimizar. Con $p$ variables hace falta álgebra matricial, y a cambio obtenemos algo que con una variable no se veía: **una fórmula cerrada**, y la respuesta exacta a cuándo existe y cuándo es única.

> **Lo que usamos de antes**
>
> **?@def-riesgo-empirico**, **?@thm-condicion-primer-orden** (la condición que se anula en el mínimo), **?@def-descenso**.

> **Repaso previo (20 minutos)**
>
> Producto matriz-vector, traspuesta, norma euclídea, y qué significa que una columna sea combinación lineal de las demás. Está en el anexo [Cálculo y álgebra](../curso/algebra.qmd), publicado antes de esta sesión y de lectura obligatoria previa.

## Preparativos

<span class="theorem-title">**Definición 1 (Matriz de diseño)**</span> $\mathbf{X}\in \mathbb{R}^{n\times (p+1)}$ con **primera columna de unos**, de modo que $x_{i0} = 1$ y el término independiente queda absorbido en $\boldsymbol{w}$. El *score* es $s_i = \mathbf{x}_i^{\mathsf{T}}\boldsymbol{w}$. Este convenio se mantiene en el resto del curso.

<span class="theorem-title">**Proposición 1 (Riesgo en forma matricial)**</span> $\displaystyle \hat{R}(\boldsymbol{w}) = \frac{1}{n}\left\lVert \mathbf{X}\boldsymbol{w}- \mathbf{y} \right\rVert^2$.

> **Dos identidades, del anexo**
>
> $\nabla(\mathbf{a}^{\mathsf{T}}\boldsymbol{w}) = \mathbf{a}$ (**?@lem-grad-lineal**) y $\nabla(\boldsymbol{w}^{\mathsf{T}}\mathbf{A}\boldsymbol{w}) = 2\mathbf{A}\boldsymbol{w}$ para $\mathbf{A}$ simétrica (**?@lem-grad-cuadratica**). Son cálculo, no aprendizaje automático: están demostradas en el anexo de [cálculo y álgebra](../curso/algebra.qmd) y se piden como trabajo previo a esta sesión.

<span class="proof-title">*Observación 1*. </span>$\mathbf{X}^{\mathsf{T}}\mathbf{X}$ es simétrica y de tamaño $(p+1)\times(p+1)$, **no** $n\times n$.

## Ecuaciones normales

<span class="theorem-title">**Teorema 1 (Gradiente del riesgo)**</span> $\displaystyle \nabla\hat{R}(\boldsymbol{w}) = \frac{2}{n}\,\mathbf{X}^{\mathsf{T}}(\mathbf{X}\boldsymbol{w}- \mathbf{y})$.

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Teorema 2 (Ecuaciones normales)**</span> Si $\hat{\boldsymbol{w}}$ minimiza $\hat{R}$, entonces $\mathbf{X}^{\mathsf{T}}\mathbf{X}\hat{\boldsymbol{w}}= \mathbf{X}^{\mathsf{T}}\mathbf{y}$.

<span class="proof-title">*Demostración*. </span>**?@thm-condicion-primer-orden** aplicado a <a href="#thm-gradiente-matricial" class="quarto-xref">Teorema 1</a>: en un mínimo el gradiente se anula, y $\frac{2}{n}\mathbf{X}^{\mathsf{T}}(\mathbf{X}\hat{\boldsymbol{w}}- \mathbf{y}) = \mathbf{0}$ equivale a $\mathbf{X}^{\mathsf{T}}\mathbf{X}\hat{\boldsymbol{w}}= \mathbf{X}^{\mathsf{T}}\mathbf{y}$.

<span class="theorem-title">**Corolario 1 (Ortogonalidad de los residuos)**</span> Con $\mathbf{r}= \mathbf{y}- \mathbf{X}\hat{\boldsymbol{w}}$ se tiene $\mathbf{X}^{\mathsf{T}}\mathbf{r}= \mathbf{0}$. En particular, **si hay término independiente**, $\sum_i r_i= 0$ y la media de las predicciones coincide con $\bar{y}$.

<span class="proof-title">*Demostración*. </span>Reescritura de <a href="#thm-ecuaciones-normales" class="quarto-xref">Teorema 2</a>. La primera componente de $\mathbf{X}^{\mathsf{T}}\mathbf{r}$ es $\mathbf{1}^{\mathsf{T}}\mathbf{r}= \sum_i r_i$ porque la primera columna de $\mathbf{X}$ es de unos.

> **Important**
>
> <a href="#cor-residuos-ortogonales" class="quarto-xref">Corolario 1</a> se utiliza en el capítulo 8 para demostrar **?@thm-sct-descompone**. Fuera de la muestra deja de cumplirse, y de ahí que $R^2$ pueda salir negativo.

## Óptimo global, existencia y unicidad

<span class="theorem-title">**Teorema 3 (Identidad fundamental)**</span> Si $\hat{\boldsymbol{w}}$ resuelve <a href="#thm-ecuaciones-normales" class="quarto-xref">Teorema 2</a>, entonces **para todo** $\boldsymbol{w}$: $$
\hat{R}(\boldsymbol{w}) = \hat{R}(\hat{\boldsymbol{w}}) + \frac{1}{n}\left\lVert \mathbf{X}(\boldsymbol{w}- \hat{\boldsymbol{w}}) \right\rVert^2 .
 \qquad(1)$$

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Corolario 2 (Todo punto crítico es mínimo global)**</span> El segundo sumando de <a href="#eq-identidad" class="quarto-xref">Ecuación 1</a> es $\geq 0$, luego $\hat{R}(\boldsymbol{w}) \geq \hat{R}(\hat{\boldsymbol{w}})$ para todo $\boldsymbol{w}$.

> **Important**
>
> Obsérvese que la optimalidad global se ha obtenido sin usar la convexidad. El argumento se reduce a sumar y restar un término.

<span class="theorem-title">**Teorema 4 (Existencia y unicidad)**</span> $\mathop{\mathrm{nuc}}(\mathbf{X}^{\mathsf{T}}\mathbf{X}) = \mathop{\mathrm{nuc}}(\mathbf{X})$. En consecuencia, $\hat{\boldsymbol{w}}$ es **único** si y solo si ninguna columna de $\mathbf{X}$ es combinación lineal de las demás (es decir $\mathop{\mathrm{rango}}\mathbf{X}= p+ 1$, lo que exige $n\geq p+ 1$), y entonces $$\hat{\boldsymbol{w}}= (\mathbf{X}^{\mathsf{T}}\mathbf{X})^{-1}\mathbf{X}^{\mathsf{T}}\mathbf{y}.$$

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Teorema 5 (Hessiano y convexidad)**</span> $\nabla^2\hat{R}= \frac{2}{n}\mathbf{X}^{\mathsf{T}}\mathbf{X}$. Para cualquier dirección $\mathbf{v}$, la función $\varphi(t) = \hat{R}(\boldsymbol{w}+ t\mathbf{v})$ cumple $\varphi''(t) = \frac{2}{n}\left\lVert \mathbf{X}\mathbf{v} \right\rVert^2 \geq 0$: **la curvatura en toda dirección es no negativa**, es decir $\hat{R}$ es convexa. Es estrictamente convexa si y solo si $\mathop{\mathrm{rango}}\mathbf{X}= p+1$.

**La comprobación queda para el laboratorio**

El cálculo es el mismo de <a href="#thm-gradiente-matricial" class="quarto-xref">Teorema 1</a> derivado una segunda vez, y la curvatura direccional vuelve a ser la línea $\left\lVert \mathbf{X}\mathbf{v} \right\rVert^2$ de <a href="#thm-existencia-unicidad" class="quarto-xref">Teorema 4</a>. No lo hacemos en pizarra porque <a href="#cor-minimo-global" class="quarto-xref">Corolario 2</a> ya nos da el mínimo global sin necesidad de convexidad; el hessiano se comprueba numéricamente en el laboratorio y se retoma en el capítulo 11, donde sí hace falta.

> **Note**
>
> Orden deliberado: <a href="#thm-identidad-fundamental" class="quarto-xref">Teorema 3</a> da el mínimo global **primero**, y el hessiano llega después como segunda lectura. Al revés, la convexidad costaría media pizarra de teoría de funciones convexas; así cuesta tres líneas. El hessiano se queda porque es lo que generaliza a *ridge* (capítulo 11) y a Newton.

## Coste computacional

<span class="theorem-title">**Proposición 2 (¿Fórmula cerrada o gradiente?)**</span> Formar $\mathbf{X}^{\mathsf{T}}\mathbf{X}$ cuesta $O(np^2)$; resolver el sistema, $O(p^3)$. Un paso de descenso cuesta $O(np)$, y $T$ pasos $O(Tnp)$. **El gradiente gana cuando $p^2 \gg Tn$**, o cuando $\mathbf{X}$ no cabe en memoria, o cuando los datos llegan en flujo.

**Resolver el sistema en lugar de invertir la matriz**

Se resuelve el sistema (`solve`, `lstsq`), no se calcula la inversa. El número de condición cumple $\kappa(\mathbf{X}^{\mathsf{T}}\mathbf{X}) = \kappa(\mathbf{X})^2$, que **enunciamos y no demostramos**: en el laboratorio se ve numéricamente con una matriz casi colineal.

## La conversación pendiente con econometría

<span class="proof-title">*Observación 2* (Aquí declinamos formalmente la inferencia). </span>En este punto la tentación es calcular errores estándar, estadísticos $t$ y valores $p$ sobre $\hat{\boldsymbol{w}}$. **No lo vamos a hacer**, y conviene decir por qué en lugar de omitirlo en silencio. Toda esa maquinaria responde a la pregunta *¿cuánto vale $w_j$?*; este curso responde a *¿qué predice mejor fuera de la muestra?*. Son objetivos distintos y llevan a modelos distintos: la regularización del capítulo 11 sesga deliberadamente $\hat{\boldsymbol{w}}$ porque a cambio reduce el error de predicción.

La deuda no queda impagada: en la semana 15 daremos incertidumbre sobre las **predicciones** mediante intervalos conformes, sin un solo valor $p$.

## Laboratorio

## Error frecuente

**`np.linalg.inv(X.T @ X) @ X.T @ y`**

Funciona con matrices bien condicionadas y puede fallar con datos reales. Dos motivos: si alguna columna es combinación de las demás, la inversa **no existe** (<a href="#thm-existencia-unicidad" class="quarto-xref">Teorema 4</a>) y NumPy devuelve basura numérica en lugar de un error; y aunque exista, elevar al cuadrado el número de condición pierde precisión. Use `solve` o `lstsq`.

## Notación ↔ código

| Matemáticas | Código |
|----|----|
| $\mathbf{X}$ con columna de unos | `np.column_stack([np.ones(n), X])` |
| $\mathbf{X}^{\mathsf{T}}\mathbf{X}\hat{\boldsymbol{w}}= \mathbf{X}^{\mathsf{T}}\mathbf{y}$ | `np.linalg.solve(X.T @ X, X.T @ y)` |
| $\hat{\boldsymbol{w}}$ | `LinearRegression(fit_intercept=False).fit(X, y).coef_` |
| $\nabla\hat{R}$ comprobado | `jax.grad(riesgo)(w, X, y)` |
| $\mathop{\mathrm{rango}}\mathbf{X}$ | `np.linalg.matrix_rank(X)` |

## Resumen

- $\mathbf{X}^{\mathsf{T}}\mathbf{X}\hat{\boldsymbol{w}}= \mathbf{X}^{\mathsf{T}}\mathbf{y}$ sale de anular el gradiente (<a href="#thm-ecuaciones-normales" class="quarto-xref">Teorema 2</a>).
- La identidad <a href="#eq-identidad" class="quarto-xref">Ecuación 1</a> da el **mínimo global** sin teoría de convexidad, y de regalo la unicidad (<a href="#thm-existencia-unicidad" class="quarto-xref">Teorema 4</a>).
- La existencia y unicidad (<a href="#thm-existencia-unicidad" class="quarto-xref">Teorema 4</a>) y la convexidad (<a href="#thm-hessiano" class="quarto-xref">Teorema 5</a>) se apoyan en la igualdad $\mathbf{v}^{\mathsf{T}}\mathbf{X}^{\mathsf{T}}\mathbf{X}\mathbf{v} = \left\lVert \mathbf{X}\mathbf{v} \right\rVert^2$.
