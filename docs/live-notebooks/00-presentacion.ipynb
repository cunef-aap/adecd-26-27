## Qué es el aprendizaje automático

Un portal de alquiler tiene 15.000 anuncios de Madrid con su precio por noche. De cada anuncio sabe cuánta gente admite, cuántos dormitorios y baños tiene, cuántas reseñas ha recibido y qué puntuación le ponen los huéspedes. Llega un anuncio nuevo, todavía sin precio. ¿Cuánto debería costar?

Nadie ha escrito la fórmula del precio de un piso en Madrid, y no parece que nadie vaya a escribirla. Lo que hay son 15.000 casos resueltos. El aprendizaje automático reúne las técnicas que construyen la regla **a partir de los casos resueltos**, en lugar de programarla a mano.

El anuncio que importa es el que todavía no tiene precio. Una regla que reprodujera los 15.000 precios conocidos y fallara en el siguiente no serviría para nada, y en el capítulo 1 comprobaremos, con estos mismos datos, que construir una regla así es fácil. Por eso en todo el curso un modelo se juzga por lo que hace con datos que **no** se han usado para construirlo.

Con el ejemplo delante ya se puede decir en general. Un problema de aprendizaje queda planteado cuando se fijan tres cosas:

1.  **Qué se observa.** Aquí, las características de un anuncio junto con su precio.
2.  **Qué se construye.** Aquí, una función que convierte características en un precio.
3.  **Con qué criterio se juzga.** Aquí, el error de esa función en anuncios que no se han usado para construirla.

Cambiando esas tres casillas se obtienen las demás familias. Dentro del caso supervisado, las casillas 2 y 3 se concretan en las tres decisiones del apartado *El esquema del curso*.

## Las tres familias

La clasificación habitual no se basa en el algoritmo, sino en las tres casillas anteriores.

**Aprendizaje supervisado.** Se observan unas características y una respuesta. Se construye una función que predice la respuesta a partir de las características. El criterio es el error de predicción en datos nuevos. Es el caso del alquiler, y el de esta asignatura.

Que el criterio esté fijado de antemano es lo que hace este caso más cómodo que los otros dos: dos personas que trabajen sobre los mismos datos y midan el error de la misma forma ordenan los modelos igual.

**Aprendizaje no supervisado.** Se observan las características, pero no hay respuesta. Aquí normalmente no se predice nada. Se construye otra cosa: una división de los anuncios en grupos, una representación con menos variables, una descripción de cómo se reparten los datos.

El problema es el criterio. Si agrupamos los 15.000 anuncios en cuatro segmentos, no hay ninguna respuesta correcta con la que comparar, porque los segmentos no estaban en los datos: los ha producido el propio análisis. Dos segmentaciones distintas pueden ser las dos defendibles. Hay que elegir un criterio externo al problema, y criterios distintos dan resultados distintos.

Decir que el aprendizaje no supervisado “predice la estructura de los datos” es una imprecisión que conviene evitar desde el principio.

**Aprendizaje por refuerzo.** Los datos no vienen dados: los genera un programa al interactuar con un entorno, de modo que sus propias decisiones cambian lo que observará después. Se construye una regla de actuación y el criterio es la recompensa acumulada.

> **Dónde estamos**
>
> Esta asignatura trata **aprendizaje supervisado con respuesta numérica**: predecir un número, como un precio, una demanda o un coste.
>
> La clasificación, que es predecir una categoría, es la asignatura hermana *Aprendizaje Automático: Clasificación* \[G249\].

**Lo que falta en la definición de arriba**

Hemos dicho “datos nuevos” sin decir de dónde salen. Detrás hay un supuesto: que los datos que tenemos y los que vendrán proceden del mismo sitio, es decir, de una misma distribución de probabilidad que no conocemos. El criterio de calidad es entonces una esperanza respecto de esa distribución, y por eso no se puede calcular, solo estimar.

Esa distinción tiene consecuencias y no conviene despacharla en una frase, así que no la usamos todavía. Se enuncia y se demuestra en el capítulo 8, que es donde hace falta para saber por qué el error medido sobre datos de prueba es una estimación honrada del error futuro, y en qué momento deja de serlo.

## Predecir no es lo mismo que explicar

Con los mismos datos y las mismas fórmulas se pueden responder dos preguntas distintas. La econometría pregunta cuánto vale un coeficiente y si es significativo; aquí preguntamos qué modelo predice mejor fuera de la muestra. El capítulo 1 compara las dos en una tabla.

La consecuencia es que llevan a modelos distintos. En el capítulo 11 veremos la regularización, que **sesga los coeficientes a propósito** porque a cambio reduce el error de predicción. Sesgar los coeficientes es un defecto desde el punto de vista de la inferencia. Aquí se acepta porque reduce el error en datos nuevos.

No cubrimos inferencia sobre los coeficientes: errores estándar, contrastes e intervalos de confianza. Es una decisión deliberada, y la asignatura de econometría es el sitio donde se tratan. El capítulo 15 sí trata la incertidumbre, pero sobre las predicciones.

## El esquema del curso

Plantear un problema de predicción son tres decisiones: qué **clase de hipótesis**, qué **pérdida** y qué **optimizador**. El curso las va rellenando en ese orden, y el índice está organizado según cuál se esté rellenando.

La segunda de las tres, la pérdida, sale de un procedimiento que se aplica quince veces:

1.  Escribir un **modelo de datos**: señal más ruido.
2.  Escribir la **verosimilitud** de ese modelo.
3.  Tomar $-\log$ y obtener una **pérdida**.
4.  Comprobar que es **convexa**, que es lo que garantiza que minimizarla tenga sentido.
5.  **Minimizarla**, en forma cerrada o por descenso de gradiente.
6.  **Medir el error** con datos que no se hayan usado para ajustar, con una métrica interpretable. Que ese número estime algo, y qué, es el capítulo 8.

Se deriva entero, a mano, una vez (capítulos 2 a 6). Después se reutiliza: *ridge* y *lasso* son el paso 3 con una distribución a priori; los splines son el paso 1 con otras variables; Poisson es el paso 1 con otro ruido; los árboles son el caso en que el paso 4 falla y el 5 se vuelve búsqueda voraz.

## Cómo es el material

No hay diapositivas. Hay tres cosas, y conviene saber para qué sirve cada una.

**Los apuntes son ejecutables.** Lo que se lee en la web es el código que produce las figuras. Las figuras de los capítulos se generan al publicarlos, ejecutando las celdas que aparecen justo encima. Si una figura está ahí, es que ese código funciona.

**Cada capítulo tiene un cuaderno con huecos.** Es el mismo fichero que los apuntes, pero con los trozos de código que se escriben en clase sustituidos por un hueco. El enlace está arriba del todo de cada capítulo y abre directamente en Google Colab, sin instalar nada. Como es el mismo fichero, **no se pueden desincronizar**: lo que se teclea en el laboratorio es exactamente lo que queda publicado como referencia.

**Las hojas de problemas** son para la S2, una por capítulo, y se publican con solución después de la sesión.

### Herramientas

- **Python**, con `numpy`, `pandas`, `matplotlib` y `scikit-learn`. `statsmodels` en el capítulo 14.
- **`jax`** en los capítulos 5 y 6, para **diferenciación automática**. No la usamos para entrenar nada: la usamos para comprobar que una derivada hecha a mano es correcta. `jax.grad(f)` devuelve una función, el gradiente de `f`, que es la traducción literal del operador $\nabla$. El capítulo 5 explica en qué se diferencia de las diferencias finitas y de la derivación simbólica, y por qué el modo inverso da las $p+1$ derivadas parciales en una sola pasada, que es lo que hace viable entrenar modelos grandes.
- **Google Colab**, para no instalar nada. Quien prefiera trabajar en local tiene el entorno en `environment.yml`.
- **Herramientas de IA generativa**. Su uso está permitido y, en el proyecto, es obligatorio documentarlo. Ver la [política de uso de IA](../curso/politica-ia.qmd).
