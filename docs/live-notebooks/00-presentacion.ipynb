# Presentación

Qué vamos a estudiar, cómo se trabaja y cómo se evalúa

## ¿Qué es el aprendizaje automático?

Un conjunto de métodos para construir, a partir de datos, funciones que predicen. La palabra que hace el trabajo es *predicen*: no describen los datos que ya tenemos, sino que asignan un valor a observaciones que todavía no hemos visto.

## Tipos de problemas

El primer corte es por lo que sabemos de los datos:

- **Aprendizaje supervisado**: cada observación viene con una respuesta conocida, y queremos predecir esa respuesta en observaciones nuevas.
- **Aprendizaje no supervisado**: no hay respuesta, y buscamos estructura.
- **Aprendizaje por refuerzo**: un agente actúa, recibe una recompensa y aprende de ella.

Dentro del supervisado, el segundo corte es por el tipo de respuesta:

- **Regresión**: la respuesta es un número. El precio de un piso, la demanda de la semana que viene, el coste de un siniestro.
- **Clasificación**: la respuesta es una categoría. Si un cliente impaga o no.

**Esta asignatura trata el aprendizaje supervisado con respuesta numérica.** La clasificación es la asignatura hermana, *Aprendizaje Automático: Clasificación* \[G249\], que comparte con esta buena parte del aparato.

## Predicción no es inferencia

Es la distinción que organiza el curso, y conviene tenerla clara desde el primer día.

|  | Econometría | Esta asignatura |
|------------------------|------------------------|------------------------|
| Pregunta | ¿cuánto vale $w_j$ y es significativo? | ¿qué $\boldsymbol{w}$ predice mejor fuera de la muestra? |
| Objeto de interés | el coeficiente | la predicción |
| Criterio | insesgadez, error estándar | error sobre datos no vistos |

Son preguntas distintas y llevan a modelos distintos. En el capítulo 8 veremos un método, la regularización, que **sesga los coeficientes a propósito** porque a cambio predice mejor. Desde la óptica de la inferencia eso sería un defecto; desde la nuestra, el objetivo.

No cubrimos inferencia sobre los coeficientes (errores estándar, contrastes, intervalos de confianza). Es una decisión deliberada. El capítulo 15 sí trata la incertidumbre, pero sobre las predicciones.

## El esquema del curso

Hay un único método, y se repite quince veces:

1.  Escribe un modelo de datos: **señal más ruido**.
2.  Escribe la **verosimilitud** de ese modelo.
3.  Toma $-\log$ y obtén una **pérdida**.
4.  **Minimízala**, en forma cerrada o por descenso de gradiente.
5.  **Valida** sin reutilizar los datos de ajuste y mide con una métrica interpretable.

Ese esquema se deriva entero, a mano, una vez (capítulos 2 a 4). Después se reutiliza: *ridge* y *lasso* son el paso 3 con un prior; los splines son el paso 1 con otras variables; Poisson es el paso 1 con otro ruido; los árboles son el caso en que el paso 4 se vuelve búsqueda voraz.

## Cómo se trabaja

Tres sesiones por semana, con papeles distintos:

|  | Sesión | Qué se hace |
|------------------------|------------------------|------------------------|
| S1 | **Teoría** | Una demostración, en pizarra. Sin ordenador |
| S2 | **Problemas** | Lápiz y papel. Empieza con *la derivación de la semana* |
| S3 | **Laboratorio** | Se completa en directo el cuaderno del capítulo |

**Se demuestra todo lo que se enuncia.** Es la diferencia principal con el curso anterior, y conviene decirlo antes de empezar: no es más difícil, es un modo de trabajo distinto. Lo poco que no demostramos aparece marcado como *fuera de alcance*, con el motivo escrito.

Cada S2 empieza con diez minutos en los que una persona reproduce en la pizarra una demostración elegida al azar **entre las de las tres últimas semanas**. La lista completa está publicada en el [índice de resultados](../curso/indice-resultados.qmd) desde el primer día.

## Evaluación

La evaluación continua es **obligatoria para todo el mundo**, incluidos repetidores y estudiantes de intercambio. Los parciales **no son liberatorios**.

| Actividad                                  | Peso     | Cuándo                 |
|--------------------------------------------|----------|------------------------|
| Examen parcial 1                           | 20 %     | semana 7, sesión S3    |
| Proyecto en colaboración con agentes de IA | 20 %     | semanas 12 y 13        |
| **Examen final**                           | **60 %** | convocatoria ordinaria |

El examen final evalúa todo el programa, lo redacta el profesor de la asignatura y requiere una **calificación mínima de 5 sobre 10** para poder aprobar. El bloque B sale de la lista publicada en el [índice de resultados](../curso/indice-resultados.qmd), que por tanto es un compromiso y no una orientación.

### Cómo sale la nota de actas

| Situación                       | Nota en actas                             |
|------------------------------------|------------------------------------|
| Examen final $\geq 5$           | 40 % continua + 60 % final                |
| Examen final $< 5$              | la nota del examen final                  |
| No se presenta al examen final  | la nota de la continua, ponderada al 40 % |
| No ha calificado ninguna prueba | No Presentado                             |

En **convocatoria extraordinaria**, la nota de evaluación continua puede sustituirse, cuando favorezca al estudiante, por la ponderación entre la continua (40 %) y el examen final de la convocatoria ordinaria (60 %).
