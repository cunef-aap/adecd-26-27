## Planteamiento

Antes de definir nada, vamos a mirar unos datos.

```python
import numpy as np
from matplotlib import pyplot as plt

rng = np.random.default_rng(1)
kw_puntos = dict(color="black", facecolors="none", s=38, alpha=0.7)

n = 40
x_a = np.sort(rng.uniform(0, 10, n))
y_a = 1.5 * x_a + 2 + rng.normal(scale=2.0, size=n)

x_b = np.sort(rng.uniform(0, 10, n))
y_b = 10 * np.sin(0.6 * x_b) + rng.normal(scale=1.2, size=n)
```

Figura 1: Dos conjuntos de datos. En los dos hay un patrón que relaciona $x$ con $y$.

```python
fig, axs = plt.subplots(1, 2, figsize=(8, 3.4))
for ax, (xx, yy, t) in zip(axs, [(x_a, y_a, "(a)"), (x_b, y_b, "(b)")]):
    ax.scatter(xx, yy, **kw_puntos)
    ax.set(xlabel="$x$", ylabel="$y$", title=t)
plt.tight_layout()
```

Los dos paneles de <a href="#fig-dos-patrones" class="quarto-xref">Figura 1</a> tienen algo en común: se ve que $x$ y $y$ están relacionadas, y se ve **cómo**. En el panel (a), $y$ crece de forma aproximadamente proporcional a $x$. En el panel (b), $y$ sube, baja y vuelve a subir. Si tapamos un punto cualquiera y preguntamos qué valor de $y$ le corresponde, casi cualquier persona da una respuesta razonable, y la da sin hacer ninguna cuenta.

Es decir: para detectar el patrón y para usarlo nos han bastado los ojos. El problema es que un ordenador no tiene ojos, y que con cinco predictores en lugar de uno tampoco los tenemos nosotros. La pregunta con la que arranca el curso es esta:

> ¿Cómo escribimos ese patrón de forma que una máquina pueda encontrarlo y usarlo?

## El patrón es una función

La respuesta que vamos a desarrollar durante quince semanas es que el patrón se escribe como una **función**. En el panel (a) la función es una recta; en el panel (b), una curva. En los dos casos es un objeto que toma un valor de $x$ y devuelve un valor de $y$.

Figura 2: El mismo patrón, escrito como una función. Es lo que queremos que la máquina encuentre.

```python
fig, axs = plt.subplots(1, 2, figsize=(8, 3.4))
rejilla = np.linspace(0, 10, 200)
for ax, (xx, yy, ff, t) in zip(
    axs,
    [(x_a, y_a, lambda z: 1.5 * z + 2, "(a)"),
     (x_b, y_b, lambda z: 10 * np.sin(0.6 * z), "(b)")],
):
    ax.scatter(xx, yy, **kw_puntos)
    ax.plot(rejilla, ff(rejilla), color="black", lw=2)
    ax.set(xlabel="$x$", ylabel="$y$", title=t)
plt.tight_layout()
```

Llamamos $x$ a lo que se conoce de antemano y $y$ a lo que se quiere averiguar. Si $f$ es la función que hemos dibujado, lo que pretendemos es que $$
\text{``}\ f(x) = \hat{y}\approx y\ \text{''}
$$ donde $\hat{y}$ es lo que el modelo predice e $y$ es el valor verdadero. Las comillas están puestas a propósito: eso todavía no es una afirmación que se pueda demostrar, porque no hemos dicho qué quiere decir $\approx$. Lo diremos, y el capítulo 2 lo convierte en algo preciso, al escribir la diferencia entre $\hat{y}$ e $y$ como un término de ruido con distribución conocida. De momento sirve para fijar los símbolos.

En <a href="#fig-dos-funciones" class="quarto-xref">Figura 2</a> los puntos **no están sobre la curva**, están alrededor. Escribir $=$ en lugar de $\approx$ sería falso, y además sería mala idea. Por qué es mala idea lo comprueba la sección siguiente, con datos reales.

## El vocabulario

Con el recorrido hecho ya podemos poner nombres. Todos aparecieron en la sección anterior, solo que en código.

<span class="theorem-title">**Definición 1 (Problema de predicción supervisada)**</span> Disponemos de $n$ observaciones $$
\mathcal{D}= \{(\mathbf{x}_i, y_i)\}_{i=1}^{n}, \qquad
\mathbf{x}_i \in \mathbb{R}^{p}, \quad y_i\in \mathbb{R}.
$$

Al vector $\mathbf{x}_i$ lo llamamos **predictores** de la observación $i$, y al número $y_i$, **respuesta**. El objetivo es construir, a partir de $\mathcal{D}$, una función $\hat{f}: \mathbb{R}^{p} \to \mathbb{R}$ tal que $\hat{f}(\mathbf{x}) \approx y$ en observaciones $(\mathbf{x}, y)$ que **no** pertenecen a $\mathcal{D}$. Al número $\hat{f}(\mathbf{x})$ lo llamamos **predicción** y lo escribimos $\hat{y}$.

En el recorrido, $p$ era 7 (las columnas de `CARACTERISTICAS`), $\mathbf{x}_i$ era una fila y $y_i$ el precio de ese anuncio. La condición *no pertenecen a $\mathcal{D}$* la cumple el anuncio que todavía no tiene precio, que es el caso que nos importa y que no está en la tabla. Apartar `X_test` es la forma de imitarlo con los datos que ya tenemos: los 7.180 anuncios de `X_ent` son los que ve el modelo y los 1.796 restantes hacen el papel de anuncios nuevos.

Nos falta un nombre más. Una función suelta no se puede buscar, porque hay demasiadas. Lo que se busca es una función dentro de una familia fijada de antemano, y eso es lo que se elegía al escribir `LinearRegression()` o `DecisionTreeRegressor(max_depth=3)`.

<span class="theorem-title">**Definición 2 (Clase de hipótesis y parámetros)**</span> Una **clase de hipótesis** $\mathcal{F}$ es un conjunto de funciones de $\mathbb{R}^{p}$ en $\mathbb{R}$ entre las cuales se busca $\hat{f}$. Es habitual que las funciones de $\mathcal{F}$ estén indexadas por un vector de **parámetros** $\boldsymbol{w}\in \mathbb{R}^{q}$, y entonces escribimos $f(\mathbf{x}; \boldsymbol{w})$, o simplemente $f(\mathbf{x})$ cuando $\boldsymbol{w}$ se sobreentiende.

Con la clase de hipótesis, buscar una función se convierte en buscar un vector de números, que es algo que un ordenador sabe hacer. La clase con la que trabajaremos hasta el capítulo 9 es la de las funciones afines: $$
\mathcal{F}= \left\{ f(\mathbf{x}) = w_0+ \sum_{j=1}^{p} w_jx_j
\ :\ \boldsymbol{w}\in \mathbb{R}^{p+1} \right\} ,
$$ que es la de `LinearRegression`. Produce el panel (a) de <a href="#fig-dos-funciones" class="quarto-xref">Figura 2</a> pero no el (b). El capítulo 9 amplía esta clase para cubrir también el panel (b), reutilizando las demostraciones de los capítulos anteriores.

## Medir lo que cuesta equivocarse

En el recorrido elegimos el error absoluto medio sin discutirlo. Esa elección tiene nombre y tiene consecuencias.

<span class="theorem-title">**Definición 3 (Pérdida y riesgo empírico)**</span> Una **función de pérdida** es una función $\ell: \mathbb{R}\times \mathbb{R}\to \mathbb{R}$ donde $\ell(\hat{y}, y)$ mide el coste de haber predicho $\hat{y}$ cuando el valor verdadero era $y$. El **riesgo empírico** de un modelo sobre un conjunto de datos es la pérdida media: $$
\hat{R}(\boldsymbol{w}) = \frac{1}{n} \sum_{i=1}^{n} \ell\bigl(f(\mathbf{x}_i; \boldsymbol{w}),\, y_i\bigr) .
$$

El error absoluto medio es el riesgo empírico con $\ell(\hat{y}, y) = \left\lvert \hat{y}- y \right\rvert$. La otra pérdida habitual es la cuadrática, $\ell(\hat{y}, y) = (\hat{y}- y)^2$, y da el error cuadrático medio.

No son intercambiables. Si predecir de menos cuesta el triple que predecir de más, ninguna de las dos mide lo que importa, y hay que escribir otra. El capítulo 3 explica de dónde sale la pérdida cuadrática, que es la que domina el curso, y el 15 vuelve sobre el caso asimétrico.

Elegida la pérdida, construir un modelo consiste en resolver $$
\hat{\boldsymbol{w}}= \mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} \hat{R}(\boldsymbol{w})
 \qquad(1)$$

que es lo que resolvieron por dentro las llamadas a `.fit()`, con la pérdida cuadrática. A resolver <a href="#eq-erm" class="quarto-xref">Ecuación 1</a> se le llama **minimización del riesgo empírico**, y es la operación que se repite en todos los capítulos que siguen.

Con una salvedad que conviene anotar ya: en los árboles, `.fit()` no resuelve <a href="#eq-erm" class="quarto-xref">Ecuación 1</a>. No busca el árbol de menor riesgo, sino que lo construye pregunta a pregunta quedándose cada vez con la que más baja el error en ese paso. El capítulo 13 explica por qué no queda otro remedio, y el capítulo 4 dice qué propiedad se pierde para que no lo haya.

El caso más sencillo se puede resolver entero a mano, y conviene hacerlo una vez para ver qué forma tiene la respuesta. Es el modelo que en el recorrido se llamaba `media`.

<span class="theorem-title">**Lema 1 (El mejor modelo constante en pérdida cuadrática)**</span> Sea $\mathcal{F}= \{f(\mathbf{x}) = c : c \in \mathbb{R}\}$ la clase de las funciones constantes y $\ell(\hat{y}, y) = (\hat{y}- y)^2$. Entonces el minimizador de <a href="#eq-erm" class="quarto-xref">Ecuación 1</a> es $c = \bar{y}$.

<span class="proof-title">*Demostración*. </span>Con esa clase y esa pérdida, el riesgo empírico es una función de una sola variable: $$
\hat{R}(c) = \frac{1}{n}\sum_{i=1}^{n} (c - y_i)^2 .
$$

Es un polinomio de grado 2 en $c$, así que es derivable, y $$
\hat{R}'(c) = \frac{2}{n}\sum_{i=1}^{n} (c - y_i)
= 2\left( c - \frac{1}{n}\sum_{i=1}^{n} y_i\right) = 2(c - \bar{y}) .
$$

La derivada se anula solo en $c = \bar{y}$. Como $\hat{R}''(c) = 2 > 0$ para todo $c$, ese punto es un mínimo, y es el único.

<a href="#lem-constante-media" class="quarto-xref">Lema 1</a> reaparece en el capítulo 3, demostrado por segunda vez y de otra forma, completando cuadrados. Allí lo que interesa no es el resultado, que ya lo tenemos, sino la identidad que aparece a mitad de camino, porque es la que descompone la suma de cuadrados en el capítulo 8.

## Las tres decisiones

Ya hemos usado todas las piezas. Recogidas, plantear un problema de predicción consiste en tomar tres decisiones, y todos los métodos del curso se obtienen eligiéndolas.

| Decisión | Pregunta que responde | En el recorrido fue | Dónde se trata |
|------------------|------------------|------------------|------------------|
| **Clase de hipótesis** $\mathcal{F}$ | ¿Entre qué funciones buscamos? | `LinearRegression()` | Caps. 2, 9, 13 |
| **Pérdida** $\ell$ | ¿Qué quiere decir que una predicción es buena? | la cuadrática, dentro de `.fit()`; el EAM solo para leer | Caps. 3, 4 |
| **Optimizador** | ¿Cómo encontramos la mejor función de $\mathcal{F}$? | `.fit()` | Caps. 5, 6 |

La primera decisión es la que se ve en las figuras: rectas o curvas. La tercera es computacional. La segunda suele darse por supuesta, y el capítulo 3 muestra qué cambia al elegir una pérdida u otra.

> Que un modelo prediga bien no significa que explique nada.

## Qué entra y qué no entra

En este curso trabajamos el caso supervisado con respuesta numérica. Es una elección, y tiene un motivo: es el caso donde el criterio de calidad está bien definido antes de empezar, como acabamos de comprobar en <a href="#sec-recorrido" class="quarto-xref">Sección 3</a>.

Queda fuera, y se dice aquí para que nadie lo busque en el índice:

- **Clasificación**, es decir respuesta categórica. Se estudia en G249. Casi todo lo de este curso se traslada cambiando la pérdida, y en el capítulo 14 se ve el mecanismo.
- **Aprendizaje no supervisado**, donde no hay $y_i$. Ahí el objetivo no es predecir sino describir la estructura de $\{\mathbf{x}_i\}$, y el criterio de calidad no viene dado por el problema, hay que elegirlo. Agrupamiento y reducción de dimensión son los casos habituales.
- **Aprendizaje por refuerzo**, donde los datos dependen de las acciones que toma el propio sistema.

La definición de aprendizaje automático que sirve para los tres casos es más general que <a href="#def-problema" class="quarto-xref">Definición 1</a>, y se la debemos a Mitchell (1997): un programa aprende de la experiencia $E$ respecto a una tarea $T$ y una medida de rendimiento $P$ si su rendimiento en $T$, medido por $P$, mejora con $E$. En nuestro caso $T$ es predecir $y$ a partir de $\mathbf{x}$, $E$ es el conjunto $\mathcal{D}$ y $P$ es el error en datos no vistos. Lo que cambia en los otros dos casos es $P$, y es lo que hace que no podamos tratarlos con las mismas herramientas.

Es la primera fila de la tabla de <a href="#sec-recorrido" class="quarto-xref">Sección 3</a> leída a medias. La pregunta correcta nunca es *cuánto se equivoca* sino *sobre qué datos se ha medido*. **Solución del <a href="#exr-decisiones" class="quarto-xref">Ejercicio 1</a>**

Clase de hipótesis: las funciones constantes, $\mathcal{F}= \{f(\mathbf{x}) = c : c \in \mathbb{R}\}$. Pérdida: la cuadrática, $\ell(\hat{y}, y) = (\hat{y}- y)^2$. Optimizador: la solución en forma cerrada, que es la que da la demostración de <a href="#lem-constante-media" class="quarto-xref">Lema 1</a>, anulando la derivada.

Con la pérdida absoluta $\ell(\hat{y}, y) = \left\lvert \hat{y}- y \right\rvert$ y la misma clase de hipótesis, el minimizador es una mediana muestral, y el optimizador consiste en ordenar los datos y tomar el valor central; con $n$ par, cualquier punto entre los dos centrales sirve. Con la misma clase de funciones, cambiar la pérdida cambia el resultado. Se demuestra en el capítulo 15, y el capítulo 3 da antes la pieza que lo conecta con un modelo de ruido.

1.  Explica en tres o cuatro frases por qué el árbol sin límite consigue 0,5 euros.
2.  Un compañero concluye que el árbol sin límite “aprendió mucho pero luego se le olvidó”. Corrige esa lectura.
3.  ¿Qué modelo elegirías para poner precio a un anuncio nuevo, y con qué argumento?

**Solución del <a href="#exr-leer-tabla" class="quarto-xref">Ejercicio 2</a>**

1.  Un árbol sin límite de profundidad sigue partiendo los datos hasta que en cada hoja quedan anuncios prácticamente idénticos, y en el límite uno solo. La predicción de esa hoja es entonces el precio de ese anuncio. Es decir, el árbol guarda una copia de los datos de entrenamiento, y por eso los reproduce casi exactamente.
2.  No hay ningún olvido: el modelo no ha cambiado entre las dos columnas. Lo que cambia son los datos sobre los que se mide. Sobre los que memorizó acierta, y sobre otros no tiene nada que decir, porque nunca extrajo un patrón, solo una lista.
3.  El árbol de profundidad 3 o la recta, que están empatados alrededor de 60 euros en los apartados. El argumento es que esa columna es la única que estima lo que ocurrirá con un anuncio nuevo, que es la situación en la que se va a usar el modelo. Entre los dos, con estos datos no hay motivo para preferir uno, y el capítulo 8 dará las herramientas para decidir si la diferencia entre 59,7 y 60,0 significa algo.

<span class="theorem-title">**Ejercicio 3 (Por qué el no supervisado es otro problema)**</span> Un banco quiere agrupar a sus clientes en cuatro segmentos a partir de diez variables de consumo. Explica, en tres o cuatro frases, por qué <a href="#def-problema" class="quarto-xref">Definición 1</a> no describe este problema y qué habría que añadir para poder decir si una segmentación es mejor que otra.

**Solución del <a href="#exr-no-supervisado" class="quarto-xref">Ejercicio 3</a>**

En <a href="#def-problema" class="quarto-xref">Definición 1</a> hay una respuesta $y_i$ observada, y el criterio de calidad es el error al predecirla. Aquí no hay ninguna variable objetivo: los segmentos no vienen dados en los datos, son el resultado del propio análisis, de manera que no existe nada con lo que comparar la salida.

Para poder ordenar dos segmentaciones hay que elegir un criterio externo al problema, por ejemplo la suma de distancias de cada cliente al centro de su grupo, o una medida de separación entre grupos. Criterios distintos dan segmentaciones distintas y ninguno se deduce del enunciado, que es la diferencia con el caso supervisado.

<span class="theorem-title">**Ejercicio 4 (Cuántos puntos hacen falta)**</span> En <a href="#fig-trampa" class="quarto-xref">Figura 3</a> se ha usado un polinomio de grado 11 para pasar por 12 puntos. Argumenta por qué el grado $n- 1$ siempre basta, y qué condición tienen que cumplir los $x_i$.

Comprueba además, ejecutando el código, qué pasa si se intenta lo mismo con los 40 puntos y grado 39. Explica por qué el resultado no contradice lo que acabas de argumentar.

**Solución del <a href="#exr-interpolacion" class="quarto-xref">Ejercicio 4</a>**

Un polinomio de grado $n-1$ tiene $n$ coeficientes. Imponer que pase por los $n$ puntos son $n$ ecuaciones lineales en esos coeficientes, con matriz de Vandermonde. Esa matriz es invertible si y solo si los $x_i$ son distintos dos a dos, y entonces el sistema tiene solución única. Si hay dos observaciones con el mismo $x_i$ y distinto $y_i$, ninguna función puede pasar por las dos, porque una función asigna un solo valor a cada entrada.

Con 40 puntos y grado 39 el error de ajuste no sale cero, sino del orden de 1. El argumento de arriba sigue siendo correcto: el polinomio interpolador **existe**. Lo que falla es calcularlo. La matriz de Vandermonde de grado 39 está tan mal condicionada que en aritmética de doble precisión no se puede resolver el sistema, y `numpy` avisa de ello. Es la primera aparición de una distinción que reaparece en el capítulo 6: que un problema tenga solución y que sepamos calcularla son dos cosas distintas.

## Resumen

- Un patrón en unos datos se escribe como una función, y buscar una función se convierte en buscar unos parámetros cuando se fija una clase de hipótesis.
- El error se mide sobre datos que no se han usado para construir el modelo. Sin esa condición, el mejor modelo es el que memoriza, y no sirve para nada.
- Plantear el problema son tres decisiones: clase de hipótesis, pérdida y optimizador. En el recorrido las tomó la librería; el resto del curso las trata una por una.
