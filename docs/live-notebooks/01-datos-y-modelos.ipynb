## Planteamiento

Antes de definir nada, vamos a mirar unos datos.

```python
import numpy as np
from matplotlib import pyplot as plt

rng = np.random.default_rng(1)
kw_puntos = dict(color="black", facecolors="none", s=38, alpha=0.7)

n = 40
x_a = np.sort(rng.uniform(0, 10, n))
y_a = 1.5 * x_a + 2 + rng.normal(scale=2.0, size=n)

x_b = np.sort(rng.uniform(0, 10, n))
y_b = 10 * np.sin(0.6 * x_b) + rng.normal(scale=1.2, size=n)
```

Figura 1: Dos conjuntos de datos. En los dos hay un patrón que relaciona $x$ con $y$.

```python
fig, axs = plt.subplots(1, 2, figsize=(8, 3.4))
for ax, (xx, yy, t) in zip(axs, [(x_a, y_a, "(a)"), (x_b, y_b, "(b)")]):
    ax.scatter(xx, yy, **kw_puntos)
    ax.set(xlabel="$x$", ylabel="$y$", title=t)
plt.tight_layout()
```

Los dos paneles de <a href="#fig-dos-patrones" class="quarto-xref">Figura 1</a> tienen algo en común: se ve que $x$ y $y$ están relacionadas, y se ve **cómo**. En el panel (a), $y$ crece de forma aproximadamente proporcional a $x$. En el panel (b), $y$ sube, baja y vuelve a subir. Si tapamos un punto cualquiera y preguntamos qué valor de $y$ le corresponde, casi cualquier persona da una respuesta razonable, y la da sin hacer ninguna cuenta.

Es decir: para detectar el patrón y para usarlo nos han bastado los ojos. El problema es que un ordenador no tiene ojos, y que con cinco predictores en lugar de uno tampoco los tenemos nosotros. La pregunta con la que arranca el curso es esta:

> ¿Cómo escribimos ese patrón de forma que una máquina pueda encontrarlo y usarlo?

## El patrón es una función

La respuesta que vamos a desarrollar durante quince semanas es que el patrón se escribe como una **función**. En el panel (a) la función es una recta; en el panel (b), una curva. En los dos casos es un objeto que toma un valor de $x$ y devuelve un valor de $y$.

Figura 2: El mismo patrón, escrito como una función. Es lo que queremos que la máquina encuentre.

```python
fig, axs = plt.subplots(1, 2, figsize=(8, 3.4))
rejilla = np.linspace(0, 10, 200)
for ax, (xx, yy, ff, t) in zip(
    axs,
    [(x_a, y_a, lambda z: 1.5 * z + 2, "(a)"),
     (x_b, y_b, lambda z: 10 * np.sin(0.6 * z), "(b)")],
):
    ax.scatter(xx, yy, **kw_puntos)
    ax.plot(rejilla, ff(rejilla), color="black", lw=2)
    ax.set(xlabel="$x$", ylabel="$y$", title=t)
plt.tight_layout()
```

Llamamos $x$ a lo que se conoce de antemano y $y$ a lo que se quiere averiguar. Si $f$ es la función que hemos dibujado, lo que pretendemos es que $$
\text{``}\ f(x) = \hat{y}\approx y\ \text{''}
$$ donde $\hat{y}$ es lo que el modelo predice e $y$ es el valor verdadero. Las comillas están puestas a propósito: eso todavía no es una afirmación que se pueda demostrar, porque no hemos dicho qué quiere decir $\approx$. Lo diremos, y el capítulo 3 lo convierte en algo preciso. De momento sirve para fijar los símbolos.

Fíjate en que en <a href="#fig-dos-funciones" class="quarto-xref">Figura 2</a> los puntos **no están sobre la curva**, están alrededor. Escribir $=$ en lugar de $\approx$ sería falso, y además sería mala idea. Por qué es mala idea es lo primero que vamos a comprobar, y no con datos inventados.

## El vocabulario

Con el recorrido hecho ya podemos poner nombres. Todos aparecieron en la sección anterior, solo que en código.

<span class="theorem-title">**Definición 1 (Problema de predicción supervisada)**</span> Disponemos de $n$ observaciones $$
\mathcal{D}= \{(\mathbf{x}_i, y_i)\}_{i=1}^{n}, \qquad
\mathbf{x}_i \in \mathbb{R}^{p}, \quad y_i\in \mathbb{R}.
$$

Al vector $\mathbf{x}_i$ lo llamamos **predictores** de la observación $i$, y al número $y_i$, **respuesta**. El objetivo es construir, a partir de $\mathcal{D}$, una función $\hat{f}: \mathbb{R}^{p} \to \mathbb{R}$ tal que $\hat{f}(\mathbf{x}) \approx y$ en observaciones $(\mathbf{x}, y)$ que **no** pertenecen a $\mathcal{D}$. Al número $\hat{f}(\mathbf{x})$ lo llamamos **predicción** y lo escribimos $\hat{y}$.

En el recorrido, $n$ era 7.180 (las filas de `X_ent`), $p$ era 7 (las columnas de `CARACTERISTICAS`), $\mathbf{x}_i$ era una fila y $y_i$ el precio de ese anuncio. La condición *no pertenecen a $\mathcal{D}$* es `X_test`.

Nos falta un nombre más. Una función suelta no se puede buscar, porque hay demasiadas. Lo que se busca es una función dentro de una familia fijada de antemano, y eso es lo que se elegía al escribir `LinearRegression()` o `DecisionTreeRegressor(max_depth=3)`.

<span class="theorem-title">**Definición 2 (Clase de hipótesis y parámetros)**</span> Una **clase de hipótesis** $\mathcal{F}$ es un conjunto de funciones de $\mathbb{R}^{p}$ en $\mathbb{R}$ entre las cuales se busca $\hat{f}$. Es habitual que las funciones de $\mathcal{F}$ estén indexadas por un vector de **parámetros** $\boldsymbol{w}\in \mathbb{R}^{q}$, y entonces escribimos $f(\mathbf{x}; \boldsymbol{w})$, o simplemente $f(\mathbf{x})$ cuando $\boldsymbol{w}$ se sobreentiende.

Con la clase de hipótesis, buscar una función se convierte en buscar un vector de números, que es algo que un ordenador sabe hacer. La clase con la que trabajaremos hasta el capítulo 9 es la de las funciones afines: $$
\mathcal{F}= \left\{ f(\mathbf{x}) = w_0+ \sum_{j=1}^{p} w_jx_j
\ :\ \boldsymbol{w}\in \mathbb{R}^{p+1} \right\} ,
$$ que es la de `LinearRegression`. Produce el panel (a) de <a href="#fig-dos-funciones" class="quarto-xref">Figura 2</a> pero no el (b). Que esa limitación es menos grave de lo que parece, y cómo se levanta sin cambiar ni una demostración, es el contenido del capítulo 9.

## Medir lo que cuesta equivocarse

En el recorrido elegimos el error absoluto medio sin discutirlo. Esa elección tiene nombre y tiene consecuencias.

<span class="theorem-title">**Definición 3 (Pérdida y riesgo empírico)**</span> Una **función de pérdida** es una función $\ell: \mathbb{R}\times \mathbb{R}\to \mathbb{R}$ donde $\ell(\hat{y}, y)$ mide el coste de haber predicho $\hat{y}$ cuando el valor verdadero era $y$. El **riesgo empírico** de un modelo sobre un conjunto de datos es la pérdida media: $$
\hat{R}(\boldsymbol{w}) = \frac{1}{n} \sum_{i=1}^{n} \ell\bigl(f(\mathbf{x}_i; \boldsymbol{w}),\, y_i\bigr) .
$$

El error absoluto medio es el riesgo empírico con $\ell(\hat{y}, y) = \left\lvert \hat{y}- y \right\rvert$. La otra pérdida habitual es la cuadrática, $\ell(\hat{y}, y) = (\hat{y}- y)^2$, y da el error cuadrático medio.

No son intercambiables. Si predecir de menos cuesta el triple que predecir de más, ninguna de las dos mide lo que importa, y hay que escribir otra. El capítulo 3 explica de dónde sale la pérdida cuadrática, que es la que domina el curso, y el 15 vuelve sobre el caso asimétrico.

Elegida la pérdida, construir un modelo consiste en resolver $$
\hat{\boldsymbol{w}}= \mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} \hat{R}(\boldsymbol{w})
 \qquad(1)$$

que es lo que hicieron por dentro las cuatro llamadas a `.fit()`. A resolver <a href="#eq-erm" class="quarto-xref">Ecuación 1</a> se le llama **minimización del riesgo empírico**, y es la operación que se repite en todos los capítulos que siguen.

## Las tres decisiones

Ya hemos usado todas las piezas. Recogidas, plantear un problema de predicción consiste en tomar tres decisiones, y todos los métodos del curso se obtienen eligiéndolas.

| Decisión | Pregunta que responde | En el recorrido fue | Dónde se trata |

> Que un modelo prediga bien no significa que explique nada.

## Qué entra y qué no entra

En este curso trabajamos el caso supervisado con respuesta numérica. Es una elección, y tiene un motivo: es el caso donde el criterio de calidad está bien definido antes de empezar, como acabamos de comprobar en <a href="#sec-recorrido" class="quarto-xref">Sección 3</a>.

Queda fuera, y se dice aquí para que nadie lo busque en el índice:

- **Clasificación**, es decir respuesta categórica. Se estudia en G249. Casi todo lo de este curso se traslada cambiando la pérdida, y en el capítulo 14 se ve el mecanismo.
- **Aprendizaje no supervisado**, donde no hay $y_i$. Ahí el objetivo no es predecir sino describir la estructura de $\{\mathbf{x}_i\}$, y el criterio de calidad no viene dado por el problema, hay que elegirlo. Agrupamiento y reducción de dimensión son los casos habituales.
- **Aprendizaje por refuerzo**, donde los datos dependen de las acciones que toma el propio sistema.

La definición de aprendizaje automático que sirve para los tres casos es más general que <a href="#def-problema" class="quarto-xref">Definición 1</a>, y se la debemos a Mitchell (1997): un programa aprende de la experiencia $E$ respecto a una tarea $T$ y una medida de rendimiento $P$ si su rendimiento en $T$, medido por $P$, mejora con $E$. En nuestro caso $T$ es predecir $y$ a partir de $\mathbf{x}$, $E$ es el conjunto $\mathcal{D}$ y $P$ es el error en datos no vistos. Lo que cambia en los otros dos casos es $P$, y es lo que hace que no podamos tratarlos con las mismas herramientas.

## Error frecuente

**“Mi modelo tiene un error de medio euro”**

Es la primera fila de la tabla de <a href="#sec-recorrido" class="quarto-xref">Sección 3</a> leída a medias. La pregunta correcta nunca es *cuánto se equivoca* sino *sobre qué datos se ha medido*. Un error de medio euro medido sobre los datos de entrenamiento es, si acaso, un motivo de sospecha: significa que el modelo tiene libertad suficiente para memorizar, y por tanto que probablemente lo ha hecho.

Cuando alguien presente un modelo sin decir sobre qué datos ha medido, la información que falta es justamente la única que importa.

Escribe cuál es la clase de hipótesis, cuál la pérdida y cuál el optimizador. Después repite el ejercicio con la mediana.

**Solución del <a href="#exr-decisiones" class="quarto-xref">Ejercicio 1</a>**

Clase de hipótesis: las funciones constantes, $\mathcal{F}= \{f(\mathbf{x}) = c : c \in \mathbb{R}\}$. Pérdida: la cuadrática, $\ell(\hat{y}, y) = (\hat{y}- y)^2$. Con ellas, $$
\hat{R}(c) = \frac{1}{n}\sum_i (c - y_i)^2 ,
$$ que es un polinomio de grado 2 en $c$ con derivada $\frac{2}{n}\sum_i (c - y_i)$. Igualando a cero, $c = \bar{y}$. El optimizador es la solución en forma cerrada.

Con la pérdida absoluta $\ell(\hat{y}, y) = \left\lvert \hat{y}- y \right\rvert$ y la misma clase de hipótesis, el minimizador es la mediana. Misma familia de funciones, respuesta distinta: la pérdida no es un detalle técnico, cambia el resultado. Se demuestra en el capítulo 8.

**Solución del <a href="#exr-no-supervisado" class="quarto-xref">Ejercicio 3</a>**

En <a href="#def-problema" class="quarto-xref">Definición 1</a> hay una respuesta $y_i$ observada, y el criterio de calidad es el error al predecirla. <span class="theorem-title">**Ejercicio 4 (Cuántos puntos hacen falta)**</span> En <a href="#fig-trampa" class="quarto-xref">Figura 3</a> se ha usado un polinomio de grado 39 para pasar por 40 puntos. Argumenta por qué el grado $n- 1$ siempre basta, y qué condición tienen que cumplir los $x_i$.

**Solución del <a href="#exr-interpolacion" class="quarto-xref">Ejercicio 4</a>**

Un polinomio de grado $n-1$ tiene $n$ coeficientes. Imponer que pase por los $n$ puntos son $n$ ecuaciones lineales en esos coeficientes, con matriz de Vandermonde. Esa matriz es invertible si y solo si los $x_i$ son distintos dos a dos, y entonces el sistema tiene solución única. Si hay dos observaciones con el mismo $x_i$ y distinto $y_i$, ninguna función puede pasar por las dos, porque una función asigna un solo valor a cada entrada.

## Resumen

- Un patrón en unos datos se escribe como una función, y buscar una función se convierte en buscar unos parámetros cuando se fija una clase de hipótesis.
- El error se mide sobre datos que no se han usado para construir el modelo. Sin esa condición, el mejor modelo es el que memoriza, y no sirve para nada.
- Plantear el problema son tres decisiones: clase de hipótesis, pérdida y optimizador. Las librerías las toman por ti; el resto del curso consiste en tomarlas tú.
