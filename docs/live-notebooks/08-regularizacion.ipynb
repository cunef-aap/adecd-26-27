## Resultados que deben aparecer

- Ridge cerrado: $\hat{\boldsymbol{w}}_{\lambda}= (\mathbf{X}^{\mathsf{T}}\mathbf{X}+ n\lambda\mathbf{I})^{-1}\mathbf{X}^{\mathsf{T}}\mathbf{y}$, invertible para todo $\lambda>0$. Sale de **?@thm-gradiente-matricial** en dos líneas.
- **Ridge = MAP con prior gaussiano; lasso = MAP con prior de Laplace.** El pago del hilo abierto en **?@def-verosimilitud**. Chodrow monta el andamiaje y nunca lo cobra.
- **Soft-thresholding con diseño ortonormal**: $w_j= \mathop{\mathrm{sign}}(w_j^{\text{MCO}})\left( \left\lvert w_j^{\text{MCO}} \right\rvert-\lambda \right)_{+}$. La única demostración honesta de por qué L1 da ceros exactos y L2 no.
