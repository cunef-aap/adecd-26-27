# Verosimilitud y máxima verosimilitud

De la verosimilitud gaussiana a los mínimos cuadrados

## Planteamiento

En el capítulo 2 supusimos ruido gaussiano y llegamos a que $y_i\,\vert\,\mathbf{x}_i \sim
\mathcal{N}(f(\mathbf{x}_i), \sigma^2)$. Aquí extraemos las consecuencias de esa hipótesis: **el error cuadrático medio no es una elección, es una consecuencia**. Y de paso estimamos $\sigma$, que Chodrow trata como constante conocida aunque en la práctica $\sigma$ es desconocido y debe estimarse.

> **Lo que usamos de antes**
>
> **?@def-senal-ruido**, **?@cor-target**, **?@def-verosimilitud**, **?@lem-log-argmax**, **?@def-mle**.

## Riesgo empírico

<span class="theorem-title">**Definición 1 (Riesgo empírico)**</span> Es **?@def-perdida** con $\ell(\hat{y}, y) = (\hat{y}- y)^2$: $$\hat{R}(\boldsymbol{w}) = \frac{1}{n}\sum_{i=1}^{n} \left(y_i- \hat{y}_i\right)^2 .$$ El factor $1/n$ forma parte de la definición y se mantiene en todo el curso.

<span class="theorem-title">**Definición 2 (Residuo y suma de cuadrados)**</span> $r_i= y_i- \hat{y}_i$, $\ \mathrm{SCR}= \sum_i r_i^2$, y por tanto $\hat{R}(\boldsymbol{w}) = \mathrm{SCR}/n$.

<span class="theorem-title">**Lema 1 (Invarianza afín del argmin)**</span> Si $a > 0$ entonces $\mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} \left( a\, h(\boldsymbol{w}) + b \right) = \mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} h(\boldsymbol{w})$.

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Lema 2 (El modelo constante óptimo en error cuadrático es la media)**</span> $\displaystyle \mathop{\mathrm{arg\,min}}_{c \in \mathbb{R}} \sum_{i=1}^{n} (y_i- c)^2 = \bar{y}$.

<span class="proof-title">*Demostración*. </span>

> **Important**
>
> <a href="#lem-media-optima" class="quarto-xref">Lema 2</a> se usa tres veces más en el curso: en el capítulo 7 (imputación por la media), en el capítulo 8 (el modelo nulo contra el que compara $R^2$) y en la semana 13 (la constante óptima dentro de una hoja de un árbol).

## De la log-verosimilitud al riesgo empírico

<span class="theorem-title">**Teorema 1 (Expansión de la log-verosimilitud)**</span> Bajo el modelo lineal-gaussiano de **?@def-lineal-gaussiano**, $$
\ell(\boldsymbol{w}, \sigma) = -\frac{n}{2}\log(2\pi\sigma^2) - \frac{n}{2\sigma^2}\, \hat{R}(\boldsymbol{w}).
 \qquad(1)$$

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Corolario 1 (Máxima verosimilitud = mínimos cuadrados)**</span> Para $\sigma$ fijo, $\mathop{\mathrm{arg\,max}}_{\boldsymbol{w}} \ell(\boldsymbol{w},\sigma) = \mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} \hat{R}(\boldsymbol{w})$.

<span class="proof-title">*Demostración*. </span>Aplicar <a href="#lem-invarianza-afin" class="quarto-xref">Lema 1</a> a $-\ell$, que es $\hat{R}$ multiplicado por $n/2\sigma^2 > 0$ más una constante que no depende de $\boldsymbol{w}$.

> **Important**
>
> El cuadrado del error no es un convenio: **se deduce de haber supuesto ruido gaussiano**. Si cambia el modelo de ruido, cambia la pérdida. Ese es el contenido de <a href="#prp-laplace-mae" class="quarto-xref">Proposición 1</a>.

<span class="proof-title">*Observación 1* (La verosimilitud es condicional). </span>$L= \prod_i p(y_i\,\vert\,\mathbf{x}_i) \cdot \prod_i p(\mathbf{x}_i)$, y el segundo factor no depende de $\boldsymbol{w}$, luego sale del $\mathop{\mathrm{arg\,max}}$ por <a href="#lem-invarianza-afin" class="quarto-xref">Lema 1</a>. **Por eso no modelizamos $\mathbf{x}$**: no porque no importe, sino porque es irrelevante para $\hat{\boldsymbol{w}}$.

## Estimar el ruido

<span class="theorem-title">**Teorema 2 (Perfilado de $\sigma$)**</span> Para $\boldsymbol{w}$ fijo, el maximizador de <a href="#eq-loglik" class="quarto-xref">Ecuación 1</a> en $\sigma> 0$ es $$
\hat{\sigma}^2(\boldsymbol{w}) = \frac{1}{n}\sum_{i=1}^{n} r_i^2 = \hat{R}(\boldsymbol{w}).
$$

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Corolario 2 (Perfilar $\sigma$ no cambia $\hat{\boldsymbol{w}}$)**</span> $\ell_{\mathrm{p}}(\boldsymbol{w}) = -\frac{n}{2}\left(\log 2\pi \hat{R}(\boldsymbol{w}) + 1\right)$ es estrictamente decreciente en $\hat{R}$, luego $\mathop{\mathrm{arg\,max}}\ell_{\mathrm{p}}= \mathop{\mathrm{arg\,min}}\hat{R}$; y $\hat{\sigma}^2 = \mathrm{SCR}/n$ evaluado en $\hat{\boldsymbol{w}}$.

> **Usos posteriores de $\hat{\sigma}^2$**
>
> Este número reaparece cuatro veces: aquí; en el capítulo 7, al devolver euros insesgados tras una transformación logarítmica; en el capítulo 10, como término de ruido irreducible de la descomposición sesgo-varianza; y en la semana 15, como referencia con la que juzgar la anchura de un intervalo.

$\hat{\sigma}^2$$\mathbb{E}\!\left[ \hat{\sigma}^2 \right] = \frac{n- p- 1}{n}\sigma^2$. Se enuncia y **no se demuestra**: requiere matriz sombrero, traza e idempotencia, veinte minutos de álgebra que no enseñan aprendizaje automático. El corrector es $n/(n-p-1)$ y lo demostraron en econometría.

## Otro ruido, otra pérdida

<span class="theorem-title">**Proposición 1 (Ruido de Laplace ⟹ error absoluto)**</span> Si $\varepsilon_i \sim \mathrm{Laplace}(0, b)$, con densidad $\frac{1}{2b}e^{-\left\lvert x \right\rvert/b}$, entonces $\ell(\boldsymbol{w}) = C - \frac{1}{b}\sum_i \left\lvert y_i- f(\mathbf{x}_i) \right\rvert$ y por tanto la estimación máximo-verosímil equivale a **minimizar el error absoluto medio**.

<span class="proof-title">*Demostración*. </span>

<span class="proof-title">*Observación 3* (Catálogo ruido → pérdida). </span>

| Ruido | Pérdida | Predice | Dónde |
|------------------|------------------|------------------|------------------|
| Gaussiano | cuadrática ($\mathrm{ECM}$) | la **media** | aquí |
| Laplace | absoluta ($\mathrm{EAM}$) | la **mediana** | aquí; demostrado en la semana 15 |
| Poisson | desvianza | un conteo | capítulo 14 |
| Gamma | desvianza | un importe positivo | capítulo 14 |

## Error frecuente

**“Uso RMSE porque es lo estándar”**

Elegir una métrica equivale a elegir un modelo de ruido, y por tanto elegir qué funcional de la distribución condicional se está estimando. Si el objetivo es asimétrico o tiene colas pesadas, minimizar el $\mathrm{ECM}$ estima la media condicional, que puede no ser el funcional de interés.

## Notación ↔ código

| Matemáticas               | Código                                      |
|---------------------------|---------------------------------------------|
| $\hat{R}(\boldsymbol{w})$ | `mean_squared_error(y, yhat)`               |

## Resumen

- $\ell= C - \frac{n}{2\sigma^2}\hat{R}$: los mínimos cuadrados se deducen de la hipótesis de ruido gaussiano (<a href="#thm-loglik-mse" class="quarto-xref">Teorema 1</a>, <a href="#cor-mle-minimos-cuadrados" class="quarto-xref">Corolario 1</a>).
- $\sigma$ no es un dato del problema: se estima, y vale $\hat{\sigma}^2 = \mathrm{SCR}/n$ (<a href="#thm-perfilado-sigma" class="quarto-xref">Teorema 2</a>).
- Cambiar el modelo de ruido cambia la pérdida (<a href="#prp-laplace-mae" class="quarto-xref">Proposición 1</a>). La receta es una máquina, no un caso particular.
