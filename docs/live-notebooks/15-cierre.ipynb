## Resultados que deben aparecer

- **La predicción puntual óptima depende de la pérdida.** El minimizador de $\mathbb{E}\!\left[ (Y-c)^2 \right]$ es la media, el de $\mathbb{E}\!\left[ \left\lvert Y-c \right\rvert \right]$ es la mediana y el de la pérdida *pinball* $\rho_{\tau}$ es el cuantil $\tau$. Tres pérdidas, tres funcionales distintos de la misma distribución. Completa **?@lem-media-optima** y **?@prp-laplace-mae**, y explica a posteriori por qué el capítulo 8 insistía en elegir la métrica antes de ajustar.
- **Costes asimétricos**: si quedarse corto cuesta $c$ veces lo que pasarse, el óptimo es el cuantil $c/(1+c)$. Es el resultado que convierte una conversación de negocio en un parámetro.
