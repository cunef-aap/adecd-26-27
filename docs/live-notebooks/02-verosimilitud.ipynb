# Verosimilitud y máxima verosimilitud

Mínimos cuadrados deja de ser un convenio y pasa a ser un teorema

## Por qué estamos aquí

En el capítulo 1 supusimos ruido gaussiano y llegamos a que $y_i\,\vert\,\mathbf{x}_i \sim \mathcal{N}(f(\mathbf{x}_i), \sigma^2)$. Aquí cobramos esa hipótesis: **el error cuadrático medio no es una elección, es una consecuencia**. Y de paso estimamos $\sigma$, que Chodrow trata como constante conocida y una cohorte de ADE no va a aceptar que lo sea.

>
> **?@def-senal-ruido**, **?@cor-target**, **?@def-verosimilitud**, **?@lem-log-argmax**, **?@def-mle**.

## Riesgo empírico

<span class="theorem-title">**Definición 1 (Riesgo empírico) **</span>$$\hat{R}(\boldsymbol{w}) = \frac{1}{n}\sum_{i=1}^{n} \left(y_i- \hat{y}_i\right)^2 .$$ El factor $1/n$ **forma parte de la definición** y no se negocia en todo el curso.

<span class="theorem-title">**Definición 2 (Residuo y suma de cuadrados) **</span>$r_i= y_i- \hat{y}_i$, $\ \mathrm{SCR}= \sum_i r_i^2$, y por tanto $\hat{R}(\boldsymbol{w}) = \mathrm{SCR}/n$.

<span class="theorem-title">**Lema 1 (Invarianza afín del argmin) **</span>Si $a > 0$ entonces $\mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} \left( a\, h(\boldsymbol{w}) + b \right) = \mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} h(\boldsymbol{w})$.

<span class="proof-title">*Prueba*. </span>

$\square$

<span class="theorem-title">**Lema 2 (El mejor modelo constante es la media) **</span>$\displaystyle \mathop{\mathrm{arg\,min}}_{c \in \mathbb{R}} \sum_{i=1}^{n} (y_i- c)^2 = \bar{y}$.

<span class="proof-title">*Prueba*. </span>

$\square$

> **Important**
>
> [Lema 2](#lem-media-optima) se usa **tres veces más** en el curso: en el capítulo 5 (imputación por la media), en el capítulo 6 (el modelo nulo contra el que compara $R^2$) y en la semana 11 (la constante óptima dentro de una hoja de un árbol). Es la inversión más rentable del capítulo.

## El teorema central

<span class="theorem-title">**Teorema 1 (Expansión de la log-verosimilitud) **</span>Bajo el modelo lineal-gaussiano de **?@def-lineal-gaussiano**, $$
\ell(\boldsymbol{w}, \sigma) = -\frac{n}{2}\log(2\pi\sigma^2) - \frac{n}{2\sigma^2}\, \hat{R}(\boldsymbol{w}).
 \qquad(1)$$

<span class="proof-title">*Prueba*. </span>

$\square$

<span class="theorem-title">**Corolario 1 (Máxima verosimilitud = mínimos cuadrados) **</span>Para $\sigma$ fijo, $\mathop{\mathrm{arg\,max}}_{\boldsymbol{w}} \ell(\boldsymbol{w},\sigma) = \mathop{\mathrm{arg\,min}}_{\boldsymbol{w}} \hat{R}(\boldsymbol{w})$.

<span class="proof-title">*Prueba*. </span>Aplicar [Lema 1](#lem-invarianza-afin) a $-\ell$, que es $\hat{R}$ multiplicado por $n/2\sigma^2 > 0$ más una constante que no depende de $\boldsymbol{w}$. $\square$

> **Dígalo en voz alta**
>
> Nadie decidió usar el cuadrado del error. **El cuadrado se deduce de haber supuesto ruido gaussiano.** Cambien el ruido y cambia la pérdida: eso es [Proposición 1](#prp-laplace-mae).

## La verosimilitud es condicional

$L= \prod_i p(y_i\,\vert\,\mathbf{x}_i) \cdot \prod_i p(\mathbf{x}_i)$, y el segundo factor no depende de $\boldsymbol{w}$, luego sale del $\mathop{\mathrm{arg\,max}}$ por [Lema 1](#lem-invarianza-afin). **Por eso no modelizamos $\mathbf{x}$**: no porque no importe, sino porque es irrelevante para $\hat{\boldsymbol{w}}$.

## Estimar el ruido

<span class="theorem-title">**Teorema 2 (Perfilado de $\sigma$) **</span>Para $\boldsymbol{w}$ fijo, el maximizador de [Ecuación 1](#eq-loglik) en $\sigma> 0$ es $$
\hat{\sigma}^2(\boldsymbol{w}) = \frac{1}{n}\sum_{i=1}^{n} r_i^2 = \hat{R}(\boldsymbol{w}).
$$

<span class="proof-title">*Prueba*. </span>

$\square$

<span class="theorem-title">**Corolario 2 (Perfilar $\sigma$ no cambia $\hat{\boldsymbol{w}}$) **</span>$\ell_{\mathrm{p}}(\boldsymbol{w}) = -\frac{n}{2}\left(\log 2\pi \hat{R}(\boldsymbol{w}) + 1\right)$ es estrictamente decreciente en $\hat{R}$, luego $\mathop{\mathrm{arg\,max}}\ell_{\mathrm{p}}= \mathop{\mathrm{arg\,min}}\hat{R}$; y $\hat{\sigma}^2 = \mathrm{SCR}/n$ evaluado en $\hat{\boldsymbol{w}}$.

> **El hilo de $\hat{\sigma}^2$**
>
> Este número reaparece **cuatro veces**: aquí; en el capítulo 5 al devolver euros insesgados tras una transformación logarítmica; en el capítulo 7 como el término de ruido irreducible de la descomposición sesgo-varianza; y en la semana 15 como el suelo contra el que se juzga la anchura de un intervalo. Denle un nombre y guárdenlo.

## $\hat{\sigma}^2$ está sesgado a la baja

$\mathbb{E}\!\left[ \hat{\sigma}^2 \right] = \frac{n- p- 1}{n}\sigma^2$. Se enuncia y **no se demuestra**: requiere matriz sombrero, traza e idempotencia, veinte minutos de álgebra que no enseñan aprendizaje automático. El corrector es $n/(n-p-1)$ y lo demostraron en econometría.

## Otro ruido, otra pérdida

<span class="theorem-title">**Proposición 1 (Ruido de Laplace ⟹ error absoluto) **</span>Si $\varepsilon_i \sim \mathrm{Laplace}(0, b)$, con densidad $\frac{1}{2b}e^{-\left\lvert x \right\rvert/b}$, entonces $\ell(\boldsymbol{w}) = C - \frac{1}{b}\sum_i \left\lvert y_i- f(\mathbf{x}_i) \right\rvert$ y por tanto la estimación máximo-verosímil equivale a **minimizar el error absoluto medio**.

<span class="proof-title">*Prueba*. </span>

$\square$

## Catálogo ruido → pérdida

| Ruido     | Pérdida                     | Predice             | Dónde                            |
|------------------|------------------|------------------|------------------|
| Gaussiano | cuadrática ($\mathrm{ECM}$) | la **media**        | aquí                             |
| Laplace   | absoluta ($\mathrm{EAM}$)   | la **mediana**      | aquí; demostrado en la semana 15 |
| Poisson   | desvianza                   | un conteo           | capítulo 14                      |
| Gamma     | desvianza                   | un importe positivo | capítulo 14                      |

## La trampa de esta semana

## «Uso RMSE porque es lo estándar»

Elegir métrica **es** elegir modelo de ruido, y por tanto elegir qué funcional de la distribución condicional se está estimando. Si el objetivo es asimétrico o tiene colas pesadas, minimizar $\mathrm{ECM}$ estima la media de algo cuya media no interesa.

## Notación ↔ código

| Matemáticas               | Código                                      |
|---------------------------|---------------------------------------------|
| $\hat{R}(\boldsymbol{w})$ | `mean_squared_error(y, yhat)`               |
| $\mathrm{SCR}$            | `((y - yhat)**2).sum()`                     |
| $\hat{\sigma}^2$          | `resid.var(ddof=0)` — ojo, `ddof=0`, no `1` |
| $\ell$                    | `dist.log_prob(y).sum()`                    |

## Qué llevarte

-   $\ell= C - \frac{n}{2\sigma^2}\hat{R}$: **mínimos cuadrados es un teorema** ([Teorema 1](#thm-loglik-mse), [Corolario 1](#cor-mle-minimos-cuadrados)).
-   $\sigma$ no es un dato del problema: se estima, y vale $\hat{\sigma}^2 = \mathrm{SCR}/n$ ([Teorema 2](#thm-perfilado-sigma)).
-   Cambiar el modelo de ruido cambia la pérdida ([Proposición 1](#prp-laplace-mae)). La receta es una máquina, no un caso particular.
