# De los datos crudos a la matriz de diseño

Preprocesado, datos faltantes, escalado, variables categóricas y transformación del objetivo

## Planteamiento

Hasta ahora $\mathbf{X}$ nos venía dada. En la práctica $\mathbf{X}$ **se construye**, y casi todas las decisiones de esa construcción tienen consecuencias matemáticas exactas. Este capítulo demuestra cinco.

> **Lo que usamos de antes**
>
> **?@lem-media-optima**, **?@thm-perfilado-sigma**, **?@thm-existencia-unicidad**, **?@thm-identidad-fundamental**, **?@prp-caso-cuadratico**.

## El *pipeline* es parte del modelo

<span class="theorem-title">**Definición 1 (Una transformación ajustada con datos es parte del modelo)**</span> Un preprocesado es una función $g_{\mathcal{D}}: \mathbb{R}^{p} \to \mathbb{R}^{p'}$ que **depende del conjunto $\mathcal{D}$ con el que se ajustó**: medias, escalas, niveles observados, valores de imputación. No es una operación neutra sobre la tabla.

<span class="theorem-title">**Definición 2 (Fuga de información)**</span> Hay **fuga** cuando el conjunto usado para ajustar $g$ interseca el conjunto usado para evaluar. Entonces falla la hipótesis (i) de **?@thm-test-insesgado** y el error de test deja de ser insesgado.

## Datos que faltan

<span class="theorem-title">**Definición 3 (MCAR, MAR, MNAR)**</span>  

<span class="theorem-title">**Proposición 1 (Imputar por la media contrae la varianza)**</span> Si se observan $n_{\text{obs}}$ de $n$ entradas y el resto se sustituye por $\bar{x}_{\text{obs}}$, la varianza muestral de la columna imputada es $\frac{n_{\text{obs}}}{n}\, s^2_{\text{obs}}$.

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Proposición 2 (Media más indicador = media del grupo)**</span> Con $p=1$, imputar por la media **y añadir la variable indicadora de ausencia** hace que mínimos cuadrados asigne a las filas con dato faltante exactamente su propia media $\bar{y}_{\text{falta}}$.

<span class="proof-title">*Demostración*. </span>

## Escalado

<span class="theorem-title">**Proposición 3 (Mínimos cuadrados es invariante al reescalado)**</span> Si $\tilde{\mathbf{X}} = \mathbf{X}\mathbf{D}$ con $\mathbf{D}$ diagonal invertible, entonces $\tilde{\boldsymbol{w}} = \mathbf{D}^{-1}\hat{\boldsymbol{w}}$ y **las predicciones son idénticas**.

<span class="proof-title">*Demostración*. </span>Sustituir en **?@thm-ecuaciones-normales**.

> **Important**
>
> Para **mínimos cuadrados**, estandarizar es un asunto *numérico* y no estadístico: no cambia las predicciones. En *ridge* (capítulo 11) sí pasa a ser estadístico, porque la penalización no es invariante a escala. Conviene tener presente esa diferencia antes de estandarizar por costumbre.

<span class="theorem-title">**Proposición 4 (Estandarizar y velocidad de convergencia)**</span> Con columnas ortogonales y centradas, $\hat{R}$ se desacopla en cuadráticas unidimensionales $\hat{R}_j(w_j) = a_j (w_j - w_j^\star)^2 + \text{cte}$ de curvatura $a_j = \frac{1}{n}\sum_i x_{ij}^2$. Por **?@prp-caso-cuadratico** la estabilidad exige $\alpha< 1/a_{\max}$, mientras que la dirección lenta contrae al ritmo $\left\lvert 1 - 2\alpha a_{\min} \right\rvert$: el número de iteraciones crece como $a_{\max}/a_{\min} = (s_{\max}/s_{\min})^2$. Estandarizadas, $a_1 = a_2$ y el método converge **en un paso** con $\alpha= 1/2a$.

<span class="proof-title">*Demostración*. </span>Aplicar **?@prp-caso-cuadratico** a cada coordenada por separado.

**El caso general** Lo mismo se enuncia con el número de condición de $\mathbf{X}^{\mathsf{T}}\mathbf{X}$ (cociente de autovalores). Se enuncia y se dibuja con elipses alargadas; no se demuestra.

## Variables categóricas

<span class="theorem-title">**Teorema 1 (La trampa de la variable ficticia es un teorema de rango)**</span> Si se codifican $K$ niveles con $K$ indicadoras $d_1,\dots,d_K$ **y además** hay columna de unos, entonces $\sum_k d_k = \mathbf{1}$: las columnas son linealmente dependientes, luego $\mathop{\mathrm{rango}}\mathbf{X}\le p$ y, por **?@thm-existencia-unicidad**, **el mínimo no es único**.

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Corolario 1 (Predicciones únicas, coeficientes no)**</span> El conjunto de soluciones es $\{\hat{\boldsymbol{w}}+ t\mathbf{v} : t \in \mathbb{R}\}$ y **todas producen las mismas predicciones**. Interpretar los coeficientes es imposible; predecir no se ve afectado. `drop='first'` o *ridge* restauran la unicidad.

<span class="proof-title">*Demostración*. </span>De **?@eq-identidad**: $\hat{R}(\hat{\boldsymbol{w}}+ t\mathbf{v}) = \hat{R}(\hat{\boldsymbol{w}})$ porque $\mathbf{X}\mathbf{v} = \mathbf{0}$.

> **Important**
>
> Estos dos resultados ilustran la diferencia entre predicción e inferencia sobre un caso concreto, el de un `OneHotEncoder`: el mismo modelo, con la misma capacidad predictiva, y con infinitos vectores de coeficientes distintos. Una interpretación económica de esos coeficientes sería, por tanto, la interpretación de un número arbitrario.

## Transformar el objetivo

<span class="theorem-title">**Proposición 5 (Tomar logaritmos es cambiar el modelo de ruido)**</span> $\log y = f(\mathbf{x}) + \varepsilon$ equivale a $y = e^{f(\mathbf{x})}\cdot \eta$ con $\eta = e^{\varepsilon}$ lognormal: el ruido aditivo en logaritmos es **proporcional** en euros.

<span class="theorem-title">**Teorema 2 (Sesgo de retransformación)**</span> Si $Z \sim \mathcal{N}(\mu, \sigma^2)$ entonces $\mathbb{E}\!\left[ e^{Z} \right] = e^{\mu + \sigma^2/2}$.

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Corolario 2 ($e^{\hat{y}}$ es la mediana, no la media)**</span> Las transformaciones crecientes preservan cuantiles, luego $\mathop{\mathrm{mediana}}(e^Z) = e^{\mu}$. Devolver euros insesgados **en media** exige multiplicar por $e^{\hat{\sigma}^2/2}$, con $\hat{\sigma}^2$ de **?@thm-perfilado-sigma**. Como $e^{\hat{\sigma}^2/2} > 1$, el error de exponenciar sin corregir es **sistemático y siempre a la baja**.

> **Note**
>
> $\hat{\sigma}^2$ vuelve a aparecer aquí, ahora con efecto directo sobre la predicción en euros:: un modelo de precios que exponencia sin corregir **infravalora todas** las predicciones, siempre en la misma dirección.

<span class="theorem-title">**Proposición 6 (Mínimos cuadrados en logaritmos ≈ error relativo)**</span> $\log(y/\hat{y}) \approx (y - \hat{y})/\hat{y}$ para errores pequeños, luego minimizar el error cuadrático en logaritmos equivale aproximadamente a minimizar el error **relativo**. *Esa* es la razón para tomar logaritmos de un precio; no “normalizar”.

## Valores extremos

<span class="proof-title">*Observación 1* (Atípicos y el modelo de ruido). </span>La influencia de una observación en $\hat{R}$ es $r_i^2/n$, **no acotada**; bajo error absoluto es $\left\lvert r_i \right\rvert/n$. Eliminar atípicos es **cambiar los datos**; **?@prp-laplace-mae** es **cambiar el modelo**. La segunda alternativa es poco habitual.

## Error frecuente

**`StandardScaler().fit_transform(X)` antes de `train_test_split`**

Es un error frecuente y con consecuencias importantes en el proyecto. La media y la desviación típica del escalador se han calculado **con los datos de test**, así que el error de test ya no estima nada (<a href="#def-fuga" class="quarto-xref">Definición 2</a>). Se ve poco: el modelo simplemente parece un poco mejor de lo que es.

## Resumen

- El preprocesado es **parte del modelo** (<a href="#def-pipeline" class="quarto-xref">Definición 1</a>); ajustarlo con datos de test es fuga (<a href="#def-fuga" class="quarto-xref">Definición 2</a>).
- La trampa de la variable ficticia no es una regla práctica: es **?@thm-existencia-unicidad** aplicado a *one-hot* (<a href="#thm-trampa-dummy" class="quarto-xref">Teorema 1</a>).
- Exponenciar una predicción hecha en logaritmos sin corregir por $e^{\hat{\sigma}^2/2}$ infravalora **siempre** (<a href="#cor-mediana-no-media" class="quarto-xref">Corolario 2</a>).
