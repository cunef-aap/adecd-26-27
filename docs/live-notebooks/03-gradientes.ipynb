## Derivadas parciales y gradiente

<span class="theorem-title">**Definición 1 (Derivada parcial) **</span>$\displaystyle \frac{\partial f}{\partial w_j}(\boldsymbol{w}) = \lim_{h \to 0} \frac{f(\boldsymbol{w}+ h\,\mathbf{e}_j) - f(\boldsymbol{w})}{h}$.

<span class="theorem-title">**Proposición 1 (Receta de cálculo) **</span>Para derivar respecto de $w_j$, tratar las demás coordenadas como constantes y aplicar cálculo de una variable.

<span class="theorem-title">**Definición 2 (Gradiente) **</span>$\nabla f(\boldsymbol{w}) = \left( \frac{\partial f}{\partial w_0}, \dots, \frac{\partial f}{\partial w_{p}} \right)^{\mathsf{T}}$.

<span class="theorem-title">**Proposición 2 (Aproximación de primer orden) **</span>$f(\boldsymbol{w}+ h\mathbf{u}) = f(\boldsymbol{w}) + h\, \mathbf{u}^{\mathsf{T}}\nabla f(\boldsymbol{w}) + o(h)$.

<span class="theorem-title">**Teorema 1 (Dirección de máximo crecimiento) **</span>Entre los $\mathbf{u}$ con $\left\lVert \mathbf{u} \right\rVert = 1$, la derivada direccional $\mathbf{u}^{\mathsf{T}}\nabla f$ se maximiza en $\mathbf{u}^\star = \nabla f/ \left\lVert \nabla f \right\rVert$.

<span class="proof-title">*Prueba*. </span>

$\square$

## Puntos críticos

<span class="theorem-title">**Definición 3 (Punto crítico, mínimo local y global) **</span>$\boldsymbol{w}^\star$ es punto crítico si $\nabla\hat{R}(\boldsymbol{w}^\star) = \mathbf{0}$; es mínimo local si existe $r>0$ con $\hat{R}(\boldsymbol{w}^\star) \le \hat{R}(\boldsymbol{w})$ para todo $\left\lVert \boldsymbol{w}- \boldsymbol{w}^\star \right\rVert < r$; global si vale para todo $\boldsymbol{w}$.

<span class="theorem-title">**Teorema 2 (Condición de primer orden) **</span>Si $\hat{R}$ es diferenciable y $\boldsymbol{w}^\star$ es mínimo local, entonces $\nabla\hat{R}(\boldsymbol{w}^\star) = \mathbf{0}$. **El recíproco es falso**: $f(w)=w^3$ en $w=0$.

<span class="proof-title">*Prueba*. </span>

$\square$

## El gradiente del riesgo

<span class="theorem-title">**Teorema 3 (Gradiente del riesgo, caso $p= 1$) **</span>Con $\hat{y}_i= w_0+ w_1 x_i$, $$
\frac{\partial \hat{R}}{\partial w_0} = -\frac{2}{n}\sum_i (y_i- w_0- w_1 x_i),
\qquad
\frac{\partial \hat{R}}{\partial w_1} = -\frac{2}{n}\sum_i x_i(y_i- w_0- w_1 x_i).
$$

<span class="proof-title">*Prueba*. </span>

$\square$

## Descenso de gradiente

<span class="theorem-title">**Definición 4 (Descenso de gradiente) **</span>$$\boldsymbol{w}^{(t+1)} = \boldsymbol{w}^{(t)} - \alpha\, \nabla\hat{R}\!\left(\boldsymbol{w}^{(t)}\right).$$

> **Warning**
>
> Chodrow tiene una errata en su definición de *momentum* (Def. 11.2): escribe $\boldsymbol{w}^{(t+1)}$ a ambos lados. No la hereden si algún día lifteamos ese capítulo.

<span class="theorem-title">**Proposición 3 (El descenso desciende) **</span>Si $\nabla\hat{R}(\boldsymbol{w}) \neq \mathbf{0}$, existe $\bar{\alpha} > 0$ tal que $\hat{R}(\boldsymbol{w}- \alpha\nabla\hat{R}(\boldsymbol{w})) < \hat{R}(\boldsymbol{w})$ para todo $\alpha\in (0, \bar{\alpha})$.

<span class="proof-title">*Prueba*. </span>

$\square$

<span class="theorem-title">**Proposición 4 (Caso cuadrático exacto: los tres regímenes) **</span>Si $\hat{R}(w) = a(w - w^\star)^2$ con $a > 0$, entonces $$w^{(t)} - w^\star = (1 - 2\alpha a)^t \left( w^{(0)} - w^\star \right).$$ Converge si y solo si $0 < \alpha< 1/a$; monótonamente si $\alpha< 1/2a$; oscilando si $1/2a < \alpha< 1/a$; diverge si $\alpha> 1/a$.

<span class="proof-title">*Prueba*. </span>

$\square$

> **Important**
>
> [Proposición 4](#prp-caso-cuadratico) no es una curiosidad: es el andamio con el que el **capítulo 5** justifica la estandarización sin necesidad de autovalores. No la corten.
