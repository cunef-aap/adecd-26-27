## Derivadas parciales y gradiente

<span class="theorem-title">**Definición 1 (Derivada parcial)**</span> $\displaystyle \frac{\partial f}{\partial w_j}(\boldsymbol{w}) = \lim_{h \to 0} \frac{f(\boldsymbol{w}+ h\,\mathbf{e}_j) - f(\boldsymbol{w})}{h}$.

<span class="theorem-title">**Proposición 1 (Receta de cálculo)**</span> Para derivar respecto de $w_j$, tratar las demás coordenadas como constantes y aplicar cálculo de una variable.

<span class="theorem-title">**Definición 2 (Gradiente)**</span> $\nabla f(\boldsymbol{w}) = \left( \frac{\partial f}{\partial w_0}, \dots, \frac{\partial f}{\partial w_{p}} \right)^{\mathsf{T}}$.

<span class="theorem-title">**Proposición 2 (Aproximación de primer orden)**</span> $f(\boldsymbol{w}+ h\mathbf{u}) = f(\boldsymbol{w}) + h\, \mathbf{u}^{\mathsf{T}}\nabla f(\boldsymbol{w}) + o(h)$.

<span class="theorem-title">**Teorema 1 (Dirección de máximo crecimiento)**</span> Entre los $\mathbf{u}$ con $\left\lVert \mathbf{u} \right\rVert = 1$, la derivada direccional $\mathbf{u}^{\mathsf{T}}\nabla f$ se maximiza en $\mathbf{u}^\star = \nabla f/ \left\lVert \nabla f \right\rVert$.

<span class="proof-title">*Demostración*. </span>

<span class="proof-title">*Observación 1* (El gradiente es ortogonal a las curvas de nivel). </span>Se enuncia y se **dibuja**; no se demuestra (requiere curvas parametrizadas y regla de la cadena multivariante). Sirve como apoyo geométrico y no se utiliza más adelante.

## Puntos críticos

<span class="theorem-title">**Definición 3 (Punto crítico, mínimo local y global)**</span> $\boldsymbol{w}^\star$ es punto crítico si $\nabla\hat{R}(\boldsymbol{w}^\star) = \mathbf{0}$; es mínimo local si existe $r>0$ con $\hat{R}(\boldsymbol{w}^\star) \le \hat{R}(\boldsymbol{w})$ para todo $\left\lVert \boldsymbol{w}- \boldsymbol{w}^\star \right\rVert < r$; global si vale para todo $\boldsymbol{w}$.

<span class="theorem-title">**Teorema 2 (Condición de primer orden)**</span> Si $\hat{R}$ es diferenciable y $\boldsymbol{w}^\star$ es mínimo local, entonces $\nabla\hat{R}(\boldsymbol{w}^\star) = \mathbf{0}$. **El recíproco es falso**: $f(w)=w^3$ en $w=0$.

<span class="proof-title">*Demostración*. </span>

## El gradiente del riesgo

<span class="theorem-title">**Teorema 3 (Gradiente del riesgo, caso $p= 1$)**</span> Con $\hat{y}_i= w_0+ w_1 x_i$, $$
\frac{\partial \hat{R}}{\partial w_0} = -\frac{2}{n}\sum_i (y_i- w_0- w_1 x_i),
\qquad
\frac{\partial \hat{R}}{\partial w_1} = -\frac{2}{n}\sum_i x_i(y_i- w_0- w_1 x_i).
$$

<span class="proof-title">*Demostración*. </span>

## Diferenciación automática

Acabamos de derivar $\nabla\hat{R}$ a mano. A partir de ahora lo haremos con frecuencia, y conviene tener una forma de comprobar si el resultado es correcto. Hay tres maneras de obtener una derivada en un ordenador, y solo una de ellas sirve.

**Diferencias finitas.** Aproximar $\frac{\partial \hat{R}}{\partial w_j} \approx
\frac{\hat{R}(\boldsymbol{w}+ h\mathbf{e}_j) - \hat{R}(\boldsymbol{w}- h\mathbf{e}_j)}{2h}$. Es fácil de programar y siempre funciona, pero tiene dos problemas: el error es de orden $h^2$ y además la resta de dos números casi iguales pierde precisión, de modo que no existe un $h$ bueno; y hace falta evaluar $\hat{R}$ dos veces por cada coordenada, es decir $2(p+1)$ evaluaciones para un solo gradiente.

**Derivación simbólica.** Manipular la expresión algebraica, como haría un humano o un sistema de álgebra computacional. Es exacta, pero la expresión crece de forma explosiva al componer funciones, y muchos programas ni siquiera tienen una expresión algebraica cerrada.

**Diferenciación automática.** Ni aproxima ni manipula símbolos. Observa que un programa que calcula $\hat{R}(\boldsymbol{w})$ es una composición de operaciones elementales cuya derivada se conoce (sumar, multiplicar, `exp`, `log`), y aplica la regla de la cadena a esa composición, **evaluándola numéricamente** paso a paso. El resultado es exacto salvo error de redondeo.

<span class="theorem-title">**Definición 4 (Modos directo e inverso)**</span> Sea $\hat{R}: \mathbb{R}^{p+1} \to \mathbb{R}$ una composición de operaciones elementales. El **modo directo** propaga derivadas desde las entradas hacia la salida y obtiene una derivada direccional por pasada. El **modo inverso** las propaga desde la salida hacia las entradas y obtiene **el gradiente completo en una sola pasada**.

Para lo que hacemos aquí, la elección está clara: nuestras funciones tienen muchas entradas ($p+1$ coeficientes) y una sola salida (un riesgo). El modo inverso da las $p+1$ derivadas parciales de una vez, a un coste comparable al de evaluar $\hat{R}$. El modo directo necesitaría $p+1$ pasadas. Por eso el modo inverso es el que usan todas las librerías de aprendizaje automático, y es lo que hace viable entrenar modelos con millones de parámetros.

En `jax`, la diferenciación automática se escribe como lo que es: un operador que toma una función y devuelve otra función, su gradiente.

## Descenso de gradiente

<span class="theorem-title">**Definición 5 (Descenso de gradiente)**</span> $$\boldsymbol{w}^{(t+1)} = \boldsymbol{w}^{(t)} - \alpha\, \nabla\hat{R}\!\left(\boldsymbol{w}^{(t)}\right).$$

> **Warning**
>
> Chodrow tiene una errata en su definición de *momentum* (Def. 11.2): escribe $\boldsymbol{w}^{(t+1)}$ a ambos lados. Conviene corregirla si se reutiliza ese capítulo.

<span class="theorem-title">**Proposición 3 (Decrecimiento del riesgo en cada paso)**</span> Si $\nabla\hat{R}(\boldsymbol{w}) \neq \mathbf{0}$, existe $\bar{\alpha} > 0$ tal que $\hat{R}(\boldsymbol{w}- \alpha\nabla\hat{R}(\boldsymbol{w})) < \hat{R}(\boldsymbol{w})$ para todo $\alpha\in (0, \bar{\alpha})$.

<span class="proof-title">*Demostración*. </span>

<span class="theorem-title">**Proposición 4 (Caso cuadrático exacto: los tres regímenes)**</span> Si $\hat{R}(w) = a(w - w^\star)^2$ con $a > 0$, entonces $$w^{(t)} - w^\star = (1 - 2\alpha a)^t \left( w^{(0)} - w^\star \right).$$ Converge si y solo si $0 < \alpha< 1/a$; monótonamente si $\alpha< 1/2a$; oscilando si $1/2a < \alpha< 1/a$; diverge si $\alpha> 1/a$.

<span class="proof-title">*Demostración*. </span>

## Error frecuente

**“No converge, subo la tasa de aprendizaje”**

Normalmente el problema es el opuesto. Por <a href="#prp-caso-cuadratico" class="quarto-xref">Proposición 4</a>, pasado $\alpha= 1/a$ el método **diverge**, y entre $1/2a$ y $1/a$ oscila mientras converge, que es justo lo que parece inestabilidad. Y si las variables tienen escalas muy distintas, no hay ninguna $\alpha$ buena para todas a la vez: eso se arregla estandarizando, no tocando $\alpha$.
